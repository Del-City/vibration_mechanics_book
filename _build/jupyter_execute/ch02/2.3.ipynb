# 多自由度系统的强迫振动

## 阻尼的表达与处理

至今仅限于讨论多自由度无阻尼系统的振动分析。如果系统的阻尼很小，而激励频率又远离共振频率，阻尼效应影响很小，可以略去不计。但当激励频率接近固有频率时，阻尼的影响显著，便不能忽略其影响。

由于阻尼机理很复杂，通常采用黏性阻尼假设，即假设系统的阻尼力大小与速度的一次方成正比，于是，系统的运动微分方程式为

$$
\boldsymbol{M} \ddot{\boldsymbol{q}} + \boldsymbol{C} \dot{\boldsymbol{q}} + \boldsymbol{K} \boldsymbol{q} = \boldsymbol{F}(t)
$$ (eq:2-94)

式中，$\boldsymbol{C}$——阻尼矩阵，与 $\boldsymbol{M}$、$\boldsymbol{K}$ 同样为对称矩阵，其元素 $C_{ij}\ (i,j=1,2,\dots,n)$ 成为阻尼影响系数。

在结构动力分析时，经常采用的是以下的瑞利阻尼假设，即阻尼矩阵 $\boldsymbol{C}$ 为如下形式：

$$
\boldsymbol{C} = \alpha \boldsymbol{M} + \beta \boldsymbol{K}
$$ (eq:2-95)

式中，$\alpha$ 与 $\beta$——比例常数。上式表示阻尼矩阵正比于质量矩阵（$\beta=0$）或正比于刚度矩阵（$\alpha=0$），或者正比于它们的线性组合，称这种阻尼为比例阻尼。

如果用无阻尼系统的固有振型，便可使 $\boldsymbol{C}$ 对角化。设 $\boldsymbol{\varPhi}$ 为正则化振型，则

$$
\boldsymbol{\varPhi}^\mathrm{T} \boldsymbol{C} \boldsymbol{\varPhi} = \alpha \boldsymbol{\varPhi}^\mathrm{T} \boldsymbol{M} \boldsymbol{\varPhi} + \beta \boldsymbol{\varPhi}^\mathrm{T} \boldsymbol{K} \boldsymbol{\varPhi} = \alpha \boldsymbol{I} + \beta \boldsymbol{\varLambda}
$$ (eq:2-96)

式中，$\boldsymbol{I}$——单位矩阵；$\boldsymbol{\varLambda}$——对角矩阵，

$$
\boldsymbol{\varLambda} = \mathrm{diag}\left( \omega_{\mathrm{n}r}^2 \right)
$$ (eq:2-97)

令

$$
\alpha + \beta \omega_{\mathrm{n}r}^2 = 2\omega_{\mathrm{n}r} \zeta_r
$$ (eq:2-98)

或写成

$$
\zeta_r = \frac{\alpha + \beta \omega_{\mathrm{n}r}^2}{2\omega_{\mathrm{n}r}} = \frac{\alpha}{2\omega_{\mathrm{n}r}} + \frac{\beta \omega_{\mathrm{n}r}}{2}
$$ (eq:2-99)

称 $\zeta_r$ 为振型（或模态）比例阻尼比。由式 {eq}`eq:2-99` 可见，当阻尼矩阵正比于质量矩阵（$\beta=0$）时，阻尼比 $\zeta_r$ 与频率 $\omega_{\mathrm{n}r}$ 成反比，因此在系统的总响应中，低阶振型起的作用较小。而反之，若阻尼矩阵正比于刚度矩阵，即 $\alpha=0$，阻尼比 $\zeta_r$ 正比于频率 $\omega_{\mathrm{n}r}$，则高阶振型起的作用较小。因此，适当地选取 $\alpha$ 与 $\beta$，可以近似地反映实际振动系统中阻尼的影响。

由于阻尼的机理很复杂，至今并不完全清楚，故精确地确定阻尼矩阵是困难的，通常采用实验测定方法近似地确定阻尼比 $\zeta_r$。由于系统并非总是有比例阻尼，因此，非对角元素不总是全为零，但当阻尼比很小时，可以令非对角元素为零，建立对角阻尼矩阵。

这样，对应正则坐标的阻尼矩阵为对角阵，

$$
\boldsymbol{C} =
\begin{bmatrix}
2\omega_{\mathrm{n}1}\zeta_1 & & & & \\
& 2\omega_{\mathrm{n}2}\zeta_2 & & & \\
& & \ddots & & \\
& & & 2\omega_{\mathrm{n}r}\zeta_r & \\
& & & & \ddots \\
& & & & & 2\omega_{\mathrm{n}n}\zeta_n
\end{bmatrix}
$$ (eq:2-100)

对于这样的阻尼矩阵，可以对方程式解耦，给求解多自由度系统的响应带来很大的便利。

## 系统对任意激励的响应

线性多自由度系统有阻尼强迫振动的运动方程式{eq}`eq:2-94`为

$$
\boldsymbol{M} \ddot{\boldsymbol{q}} + \boldsymbol{C} \dot{\boldsymbol{q}} + \boldsymbol{K} \boldsymbol{q} = \boldsymbol{F}(t)
$$ (eq:2-101)

一般情况下，$\boldsymbol{M}$、$\boldsymbol{C}$ 和 $\boldsymbol{K}$ 这三个矩阵中有非零的耦合项，如 $K_{ij}=K_{ji}\neq0$。因而需要解有 $n$ 个未知数的常微分方程组。利用广义坐标 $\boldsymbol{q}$ 和主坐标 $\boldsymbol{p}$ 之间的变换关系<!--式{eq}`eq:2-51`-->，可以将广义坐标的响应用主振型和主坐标乘积的级数和来描述，这就是展开定理，而用此求响应的方法称为振型叠加法。主振型取无阻尼时的固有振型，主坐标是待求的，从而可将用广义坐标描述的振动方程转换为用主坐标表示的振动方程表示，求出主坐标的响应式，然后再反回求出广义坐标的响应。这是一种求多自由度系统响应的有效方法。

为此，先要求解自由振动中的特征值问题：

$$
\left( \boldsymbol{K} - \omega_{\mathrm{n}r}^2 \boldsymbol{M} \right) \boldsymbol{A}_r = \boldsymbol{0}
$$ (eq:2-102)

由此确定固有振型 $\boldsymbol{A}_r$ 和固有频率 $\omega_{\mathrm{n}r}(r=1,2,\dots,n)$。然后，作坐标变换

$$
\boldsymbol{q} = \boldsymbol{\varPhi} \boldsymbol{p}
$$ (eq:2-103)

为取更普遍化的形式，上式中 $\boldsymbol{\varPhi}$ 取未正则化的振型，$\boldsymbol{p}$ 为系统的主坐标矢量。再将此变换式代入运动方程式 {eq}`eq:2-101` 并前乘 $\boldsymbol{\varPhi}^\mathrm{T}$，可得

$$
\widetilde{\boldsymbol{M}} \ddot{\boldsymbol{p}} + \widetilde{\boldsymbol{C}} \dot{\boldsymbol{p}} + \widetilde{\boldsymbol{K}} \boldsymbol{p} = \widetilde{\boldsymbol{F}}(t)
$$ (eq:2-104)

式中，

$$
\left.
\begin{aligned}
&\widetilde{\boldsymbol{M}} = \boldsymbol{\varPhi}^\mathrm{T} \boldsymbol{M} \boldsymbol{\varPhi} \text{——模态质量矩阵} \\
&\widetilde{\boldsymbol{C}} = \boldsymbol{\varPhi}^\mathrm{T} \boldsymbol{C} \boldsymbol{\varPhi} \text{——模态阻尼矩阵} \\
&\widetilde{\boldsymbol{K}} = \boldsymbol{\varPhi}^\mathrm{T} \boldsymbol{K} \boldsymbol{\varPhi} \text{——模态刚度矩阵} \\
&\widetilde{\boldsymbol{F}}(t) = \boldsymbol{\varPhi}^\mathrm{T} \boldsymbol{F}(t) \text{——模态力矢量}
\end{aligned}
\right\}
$$ (eq:2-105)

由正交条件知，$\boldsymbol{M}$ 和 $\boldsymbol{K}$ 是对角矩阵，若 $\widetilde{\boldsymbol{C}}$ 是黏性阻尼形式的对角阵，则式 {eq}`eq:2-104` 解耦为主坐标描述的振动方程：

$$
\ddot{p}_r + 2\zeta_r \omega_{\mathrm{n}r} \dot{p}_r + \omega_{\mathrm{n}r}^2 p_r = \frac{\widetilde{F}_r(t)}{M_r} \quad (r=1,2,\dots,n)
$$ (eq:2-106)

式中，

$$
M_r = \boldsymbol{\varphi}^{(r)\mathrm{T}} \boldsymbol{M} \boldsymbol{\varphi}^{(r)}, \quad \widetilde{F}_r(t) = \boldsymbol{\varphi}^{(r)\mathrm{T}} \boldsymbol{F}(t)
$$ (eq:2-107)

问题归结为求解上述单自由度系统的振动，即主振动。总响应为初始条件引起的自由振动响应与激励力响应之和：

$$
\begin{aligned}
p_r(t) &= \mathrm{e}^{-\zeta_r \omega_{\mathrm{n}r} t} \left[ p_r(0)\cos\omega_{\mathrm{d}r} t + \frac{\dot{p}_r(0) + \zeta_r \omega_{\mathrm{n}r} p_r(0)}{\omega_{\mathrm{d}r}} \sin\omega_{\mathrm{d}r} t \right] + \\
&\quad \frac{1}{\omega_{\mathrm{d}r}} \int_{0}^{t} \frac{\widetilde{F}_r(\tau)}{M_r} \mathrm{e}^{-\zeta_r \omega_{\mathrm{n}r} (t-\tau)} \sin\omega_{\mathrm{d}r} (t-\tau) \mathrm{d}\tau \quad (r=1,2,\dots,n)
\end{aligned}
$$ (eq:2-108)

若广义坐标的初始条件为 $t=0$ 时，$\boldsymbol{q}=\boldsymbol{q}(0)$，$\dot{\boldsymbol{q}}=\dot{\boldsymbol{q}}(0)$，它们与主坐标的初始条件 $\boldsymbol{p}(0)$ 和 $\dot{\boldsymbol{p}}(0)$ 之间的关系不难由式 {eq}`eq:2-103` 写出：

$$
\boldsymbol{q}(0) = \boldsymbol{\varPhi} \boldsymbol{p}(0), \quad \dot{\boldsymbol{q}}(0) = \boldsymbol{\varPhi} \dot{\boldsymbol{p}}(0)
$$ (eq:2-109)

然后前乘式 {eq}`eq:2-109` 以 $\boldsymbol{\varPhi}^\mathrm{T} \boldsymbol{M}$，得

$$
\boldsymbol{\varPhi}^\mathrm{T} \boldsymbol{M} \boldsymbol{q}(0) = \widetilde{\boldsymbol{M}} \boldsymbol{p}(0), \quad \boldsymbol{\varPhi}^\mathrm{T} \boldsymbol{M} \dot{\boldsymbol{q}}(0) = \widetilde{\boldsymbol{M}} \dot{\boldsymbol{p}}(0)
$$ (eq:2-110)

因为 $\widetilde{\boldsymbol{M}}$ 是对角阵，故由上式可得主坐标的初始条件为

$$
p_r(0) = \left( \frac{1}{M_r} \right) \boldsymbol{\varphi}^{(r)\mathrm{T}} \boldsymbol{M} \boldsymbol{q}(0), \quad \dot{p}_r(0) = \left( \frac{1}{M_r} \right) \boldsymbol{\varphi}^{(r)\mathrm{T}} \boldsymbol{M} \dot{\boldsymbol{q}}(0)
$$ (eq:2-111)

求得主坐标响应式 {eq}`eq:2-108` 后，再代入式 {eq}`eq:2-103` 便得广义坐标表示的响应表达式。

**图 2-15 两自由度系统**

解：该系统的运动方程式为

$$
\begin{bmatrix}M_1 & 0 \\0 & M_2\end{bmatrix}\begin{Bmatrix}\ddot{x}_1 \\\ddot{x}_2\end{Bmatrix}+\begin{bmatrix}C_1 + C_2 & -C_2 \\-C_2 & C_2 + C_3\end{bmatrix}\begin{Bmatrix}\dot{x}_1 \\\dot{x}_2\end{Bmatrix}+\begin{bmatrix}K_1 + K_2 & -K_2 \\-K_2 & K_2 + K_3\end{bmatrix}\begin{Bmatrix}x_1 \\x_2\end{Bmatrix}=\begin{Bmatrix}0 \\F_2(t)\end{Bmatrix}
$$

前已求得，当 $M_1 = M_2, K_1 = K_2 = K_3$ 时，系统的固有频率和固有振型分别为

$$
\omega_{\mathrm{n}1} = \sqrt{\frac{K_1}{M_1}}, \quad \omega_{\mathrm{n}2} = \sqrt{\frac{3K_1}{M_1}}
$$

和

$$
\boldsymbol{p}^{(1)} =
\begin{Bmatrix}
1 \\
1
\end{Bmatrix}, \quad
\boldsymbol{p}^{(2)} =
\begin{Bmatrix}
1 \\
-1
\end{Bmatrix}
$$

故其正则振型为

$$
\boldsymbol{\varphi}^{(1)} = \frac{1}{\sqrt{\boldsymbol{\rho}^{(1)\mathrm{T}} \boldsymbol{M} \boldsymbol{\rho}^{(1)}}}\begin{Bmatrix}1 \\1\end{Bmatrix} = \frac{1}{\sqrt{2M_1}}\begin{Bmatrix}1 \\1\end{Bmatrix}
$$

$$
\boldsymbol{\varphi}^{(2)} = \frac{1}{\sqrt{\boldsymbol{\rho}^{(2)\mathrm{T}} \boldsymbol{M} \boldsymbol{\rho}^{(2)}}}\begin{Bmatrix}1 \\-1\end{Bmatrix} = \frac{1}{\sqrt{2M_1}}\begin{Bmatrix}1 \\-1\end{Bmatrix}
$$

设 $C_1 = C_2 = C_3$，则运动方程式化为

$$
\begin{bmatrix}M_1 & 0 \\0 & M_2\end{bmatrix}\begin{Bmatrix}\ddot{x}_1 \\\ddot{x}_2\end{Bmatrix}+\begin{bmatrix}2C_1 & -C_1 \\-C_1 & 2C_1\end{bmatrix}\begin{Bmatrix}\dot{x}_1 \\\dot{x}_2\end{Bmatrix}+\begin{bmatrix}2K_1 & -K_1 \\-K_1 & 2K_1\end{bmatrix}\begin{Bmatrix}x_1 \\x_2\end{Bmatrix}=\begin{Bmatrix}0 \\F_2(t)\end{Bmatrix}
$$

令

$$
\boldsymbol{x}(t)=\boldsymbol{\varPhi}\boldsymbol{p}(t)
$$

代入上式并前乘 $\boldsymbol{\varPhi}^\mathrm{T}$ 可得

$$
\ddot{\boldsymbol{p}}+\boldsymbol{\varPhi}^\mathrm{T}\boldsymbol{C}\boldsymbol{\varPhi}\dot{\boldsymbol{p}}+\boldsymbol{\Lambda}\boldsymbol{p}=\boldsymbol{\varPhi}^\mathrm{T}\boldsymbol{F}=\widetilde{\boldsymbol{F}}(t)
$$

式中：

$$
\boldsymbol{\Lambda}=\mathrm{diag}\left(\omega_{\mathrm{n}1}^2,\omega_{\mathrm{n}2}^2\right)
$$

$$
\widetilde{\boldsymbol{C}}=\boldsymbol{\varPhi}^\mathrm{T}\boldsymbol{C}\boldsymbol{\varPhi}=\frac{C_1}{2M_1}\begin{bmatrix}1 & 1 \\1 & -1\end{bmatrix}\begin{bmatrix}2 & -1 \\-1 & 2\end{bmatrix}\begin{bmatrix}1 & 1 \\1 & -1\end{bmatrix}=\frac{C_1}{M_1}\begin{bmatrix}1 & 0 \\0 & 3\end{bmatrix}
$$

$\widetilde{\boldsymbol{C}}$ 是对角阵，因本例的阻尼矩阵 $\boldsymbol{C}$ 是比例阻尼 $\boldsymbol{C}=\alpha\boldsymbol{M}+\beta\boldsymbol{K}$ 在 $\alpha=0$ 及 $\beta=\frac{C_1}{K_1}$ 时的特例，故 $\widetilde{\boldsymbol{C}}$ 必为对角阵。于是可得

$$
2\zeta_1\omega_{\mathrm{n}1}=\frac{C_1}{M_1},\quad 2\zeta_2\omega_{\mathrm{n}2}=\frac{3C_1}{M_1}

或

$$
\zeta_1=\frac{C_1}{2M_1\omega_{\mathrm{n}1}},\quad \zeta_2=\frac{3C_1}{2M_1\omega_{\mathrm{n}2}}
$$

$$
\widetilde{\boldsymbol{F}}(t)=\boldsymbol{\varPhi}^\mathrm{T}\boldsymbol{F}=\frac{1}{\sqrt{2M_1}}\begin{bmatrix}1 & 1 \\1 & -1\end{bmatrix}\begin{Bmatrix}0 \\F_2(t)\end{Bmatrix}=\frac{1}{\sqrt{2M_1}}\begin{Bmatrix}F_2(t) \\-F_2(t)\end{Bmatrix}
$$

因而

$$
\ddot{p}_1(t)+2\zeta_1\omega_{\mathrm{n}1}\dot{p}_1+\omega_{\mathrm{n}1}^2p_1=\frac{1}{\sqrt{2M_1}}F_2(t)

$$

$$

故最终可由叠加原理求得

## 振型叠加法求系统对正弦激励的响应

若系统受到简谐激励 $\boldsymbol{F}(t)=\boldsymbol{F}_0\sin\omega t$ 作用，则式 {eq}`eq:2-106` 可写为

$$
\ddot{p}_r+2\zeta_r\omega_{\mathrm{n}r}\dot{p}_r+\omega_{\mathrm{n}r}^2p_r=\left(\frac{1}{M_r}\right)\widetilde{F}_r\sin\omega t
$$

式中，

$$
\widetilde{F}_r=\boldsymbol{\varphi}^{(r)\mathrm{T}}\boldsymbol{F}_0
$$ (eq:2-112)

如果将正弦力用复数力表示，则

$$
\ddot{\overline{p}}_r+2\zeta_r\omega_{\mathrm{n}r}\dot{\overline{p}}_r+\omega_{\mathrm{n}r}^2\overline{p}_r=\left(\frac{1}{M_r}\right)\widetilde{F}_r\mathrm{e}^{\mathrm{i}\omega t}
$$ (eq:2-113)

主坐标 $\overline{p}_r$ 的复数稳态解可写为如下形式：

$$
\overline{p}_r=\overline{H}(\omega)\widetilde{F}_r\mathrm{e}^{\mathrm{i}\omega t}
$$ (eq:2-114)

式中，$\overline{H}(\omega)$——主坐标的复频响应函数，其中第 $r$ 个元素为

$$
\overline{H}_r(\omega)=\frac{1/K_r}{(1-\gamma_r^2)+\mathrm{i}(2\zeta_r\gamma_r)},\quad K_r=M_r\omega_{\mathrm{n}r}^2\text{ 和 }\gamma_r=\frac{\omega}{\omega_{\mathrm{n}r}}
$$ (eq:2-115)

因此，物理坐标的频率响应可写为以下复数形式：

$$
\overline{\boldsymbol{q}}=\boldsymbol{\Phi}\overline{\boldsymbol{p}}(t)=\sum_{r=1}^n\boldsymbol{\varphi}^{(r)}\overline{p}_r(t)
$$

$$
=\sum_{r=1}^n\left(\frac{\boldsymbol{\varphi}^{(r)}\boldsymbol{\varphi}^{(r)\mathrm{T}}\boldsymbol{F}_0}{K_r}\right)\left[\frac{1}{(1-\gamma_r^2)+\mathrm{i}(2\zeta_r\gamma_r)}\right]\mathrm{e}^{\mathrm{i}\omega t}
$$ (eq:2-116)

因而，在坐标 $q_j$ 处受单位简谐力作用，而在坐标 $q_i$ 处的复数响应

$$
\overline{H}_{ij}(\omega)=\sum_{r=1}^n\left(\frac{\varphi_j^{(r)}\varphi_i^{(r)}}{K_r}\right)\left[\frac{1}{(1-\gamma_r^2)+\mathrm{i}(2\zeta_r\gamma_r)}\right]
$$ (eq:2-117)

称为复频响应函数，有时又称为传递函数，常常用于实验确定系统振动特性的场合。由式 {eq}`eq:2-117` 不难写出其实部和虚部。激励力是正弦形式，取式 {eq}`eq:2-116` 的虚部，便得到稳态响应的最终解：

$$
\boldsymbol{q}(t)=\sum_{r=1}^n\left(\frac{\boldsymbol{\varphi}^{(r)}\boldsymbol{\varphi}^{(r)\mathrm{T}}\boldsymbol{F}_0}{K_r}\right)\left[\frac{1}{\sqrt{(1-\gamma_r^2)^2+(2\zeta_r\gamma_r)^2}}\right]\sin(\omega t-\alpha_r)
$$ (eq:2-118)

式中，

$$
\alpha = \arctan\frac{2\zeta_r\gamma_r}{1-\gamma_r^2}
$$ (eq:2-119)

### 直接解法求系统对正弦激励的响应

当系统受到简谐激励作用时，可以用复数方法直接求解。设简谐激励力为 $F_0\mathrm{e}^{\mathrm{i}\omega t}$，则运动方程式为

$$
\boldsymbol{M}\ddot{\boldsymbol{q}}+\boldsymbol{C}\dot{\boldsymbol{q}}+\boldsymbol{K}\boldsymbol{q}=\boldsymbol{F}_0\mathrm{e}^{\mathrm{i}\omega t}
$$ (eq:2-120)

式中，$\boldsymbol{F}_0=[F_{01},F_{02},\dots,F_{0n}]^\mathrm{T}$——激励力的幅值分量。

设其稳态响应为

$$
\boldsymbol{q}(t)=\boldsymbol{q}_0\mathrm{e}^{\mathrm{i}\omega t}
$$ (eq:2-121)

代入式{eq}`eq:2-120`，并消去 $\mathrm{e}^{\mathrm{i}\omega t}$ 后，便得到矩阵运算式：

$$
(\boldsymbol{K}-\omega^2\boldsymbol{M}+\mathrm{i}\omega\boldsymbol{C})\boldsymbol{q}_0=\boldsymbol{F}_0
$$ (eq:2-122)

设矩阵

$$
\boldsymbol{G}=(\boldsymbol{K}-\omega^2\boldsymbol{M}+\mathrm{i}\omega\boldsymbol{C})^{-1}
$$ (eq:2-123)

代入上式即可得

$$
\boldsymbol{q}_0=\boldsymbol{G}\boldsymbol{F}_0
$$ (eq:2-124)

则式{eq}`eq:2-120`的稳态响应为

$$
\boldsymbol{q}(t)=\boldsymbol{G}\boldsymbol{F}_0\mathrm{e}^{\mathrm{i}\omega t}
$$ (eq:2-125)

由以上两式可知，矩阵 $\boldsymbol{G}$ 建立了系统的激励（又称输入）$\boldsymbol{F}_0$ 和位移响应（又称输出）$\boldsymbol{q}_0$ 之间的联系，因此称为传递函数矩阵。由于其元素是随激励频率 $\omega$ 变化的复函数，又因其具有柔度的量纲，故也可称为动柔度矩阵，有时亦称为复频接受率矩阵或导纳矩阵。

$\boldsymbol{G}$ 的物理意义是，由单位简谐力所引起的系统响应。其元素 $g_{ij}$ 表示在 $j$ 坐标上作用一个圆频率为 $\omega$ 的单位简谐力时，在 $i$ 坐标所引起的稳态响应（包括振幅和相位的响应）。在工程中，常由实验方法确定 $g_{ij}$，从而确定系统的固有频率。传递函数的测试与分析，广泛地应用于工程中。
