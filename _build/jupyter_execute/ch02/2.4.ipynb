# 主从系统的耦合振动

## 主从系统

任何一个复杂的多自由度系统总可以看成是由两个或若干个子系统组成。从振动角度看，最基本的子系统是一个单自由度系统——弹簧、质量和阻尼系统。通常，我们把所关心的、要进行仔细分析的那个系统称为主系统，而另一个或若干个我们并不十分感兴趣的，但对主系统振动性质有影响的子系统称为副系统或从属系统（简称从系统）。例如，船体-主机构成一复杂的振动系统，可将船体和主机作为两个子系统，它们通过机座、轴承、减振器等发生关系。在研究船体振动时，可将船体作为主系统，而把主机作为从系统。反之，在研究主机振动时，则主机作为主系统，船体作为从系统。因此，主从系统的确定是辩证的。

一般而言，各子系统的振动是相互耦合的。本节则从两个自由度的主从系统的自由振动 和强迫振动两方面，讨论从系统对主系统固有频率和响应的影响。其建立的方法和结论对于振动设计和分析以及振动控制有普遍意义。

## 主从系统的耦合作用

双质量、弹簧系统如{numref}`fig:2-19`所示，可以认为它由两个单质量、弹簧($M₁$、$K₁$和$M₂$、$K₂$)的单自由度系统构成。若$M₁-K₁$系统是我们感兴趣的，称其为主系统，而$M₂-K₂$则称为从系统。现在讨论从系统对主系统振动的影响。

主从系统的无阻尼自由振动方程、频率方程和固有频率可以方便地导出：

$$
M_1 \ddot{x}_1 + (K_1 + K_2) x_1 - K_2 x_2 = 0
$$ (eq:2-126)
$$
M_2 \ddot{x}_2 + K_2 x_2 - K_2 x_1 = 0
$$ (eq:2-127)

不难得出，其频率方程为

$$
D(\omega_{\mathrm{n}}^2) = (K_1 + K_2 - M_1 \omega_{\mathrm{n}}^2)(K_2 - M_2 \omega_{\mathrm{n}}^2) - K_2^2 = 0
$$ (eq:2-128)

由此得出固有频率为
 
$$
\omega_{\mathrm{n}1,2}^2 = \frac{K_1 + K_2}{2M_1} + \frac{K_2}{2M_2} \mp \sqrt{\frac{1}{4} \left[ \frac{K_1 + K_2}{M_1} + \frac{K_2}{M_2} \right]^2 - \frac{K_1 K_2}{M_1 M_2}}
$$ (eq:2-129)

令
 
$$
\omega_{01} = \sqrt{\frac{K_1}{M_1}}, \quad \omega_{02} = \sqrt{\frac{K_2}{M_2}}
$$ (eq:2-130)

分别为主系统和从系统单独存在时的固有频率，则主从系统的固有频率方程可改写为

$$
D(\omega_{\mathrm{n}}^2) = Z^2 - [1 + (1 + a)\xi]Z + \xi = 0
$$ (eq:2-131)

式中，$Z = \omega_{\mathrm{n}}^2 / \omega_{01}^2$——主从系统的无因次固有频率(平方值)；$a = M_2 / M_1$——从系统与主系统的质量比；$\xi = \omega_{02}^2 / \omega_{01}^2$——频率错开系数，其定义域为$[0, \infty)$。由式{eq}`eq:2-131`,可得主从系统的两个无因次频率值为

$$
Z_{1,2} = \frac{1 + (1+a)\xi \mp \sqrt{[1 + (1+a)\xi]^2 - 4\xi}}{2}
$$ (eq:2-132)

式中，

$$
Z_1 = \omega_{\mathrm{n}1}^2 / \omega_{01}^2, \quad Z_2 = \omega_{\mathrm{n}2}^2 / \omega_{02}^2
$$

$\omega_{\mathrm{n}1}$和$\omega_{\mathrm{n}2}$分别是主从系统的第一及第二阶固有频率。

首先，定性地判断主从系统固有频率的区间和变化：

当$\omega_{\mathrm{n}}^2 = 0$时 ，$D(\omega_{\mathrm{n}}^2) = K_1 K_2 > 0$

当$\omega_{\mathrm{n}}^2 = \omega_{01}^2$时 ，$D(\omega_{\mathrm{n}}^2) = -K_2 M_2 \omega_{01}^2 < 0$

当$\omega_{\mathrm{n}}^2 = \omega_{02}^2$时 ，$D(\omega_{\mathrm{n}}^2) = -K_2^2 < 0$

当$\omega_{\mathrm{n}}^2 \to \infty$时 ，$D(\omega_{\mathrm{n}}^2) > 0$

由此可见，组合系统的第一阶固有频率$\omega_{\mathrm{n}1}$，介于0和$\omega_{01}$之间，而第二阶固有频率$\omega_{\mathrm{n}2}$介于 $\omega_{02}$和$\infty$之间。因而，$\omega_{\mathrm{n}1} < \omega_{01} < \omega_{02} < \omega_{\mathrm{n}2}$。这表明，组合系统的第一阶固有频率 $\omega_{\mathrm{n}1}$ 始终比任何一个子系统（主系统或从系统）本身的固有频率低；而其第二阶固有频率则高于任何一个子系统本身的固有频率。这一结论对复杂系统的振动分析很有用。

其次，我们分析频率错开系数 $\xi$ 对主从系统固有频率的影响。若两个子系统的固有频率$\omega_{01}$和$\omega_{02}$相差很大，则 $\xi$ 趋于无穷大或为零。先讨论这两个极端情况。

当 $\xi=0$，$Z₁=0$，即$\omega_{\mathrm{n}1}^2 = 0$；$Z_2 = 1$，即$\omega_{\mathrm{n}2}^2 = \omega_{01}^2$。这相当于从系统的固有频率$\omega_{02} = 0$，主从系统之间等于不发生联系。主从系统的两个固有频率即为两个子系统的单独的固有频率。 

当$\xi \to \infty$, $Z_1 = \frac{1}{1 + a}$，即$\omega_{\mathrm{n}1}^2 = \frac{\omega_{01}^2}{1 + a}$；; $Z_2 \to \infty$, 即$\omega_{\mathrm{n}2}^2 \to \infty$。这相当于从系统的固有频率 $\omega_{02}^2 \to \infty$的情况，实际上即从系统质量 $M_2$ 与主系统质量 $M_1$ 刚性连接的结果。主从系统的耦合只相当于在主系统的质量上再增加$M_2$。将$\omega_{01}^2$和$a$的值代入$\omega_{\mathrm{n}1}^2$的表达式也可从数字上对此进行验证。

若主系统和从系统的固有频率相等，即$\xi = 1$。此时耦合作用十分显著。主从系统的固有频率由式{eq}`eq:2-132`可得

$$
Z_{1,2} = \frac{2 + a \pm \sqrt{a(4 + a)}}{2}
$$

下面再讨论质量比$a$的影响。若$a=0$，即从系统的质量$M_2$与主系统质量$M_1$相比极小而可忽略时

$$
Z_1 \approx 1, 即 \ \omega_{\mathrm{n}1} = \omega_{01}; \quad Z_2 \approx \xi, 即 \ \omega_{\mathrm{n}2} \approx \omega_{02}
$$

因而组合系统的固有频率$\omega_{\mathrm{n}1}$、$\omega_{\mathrm{n}2}$和两个子系统本身的固有频率相差甚微，故两子系统之间的耦合作用很小。随着$a$的增加，耦合系统的固有频率$\omega_{\mathrm{n}1}$、$\omega_{\mathrm{n}2}$和子系统的固有频率$\omega_{01}$、$\omega_{01}$差异也随之增大，主从系统的耦合作用越来越强。

主从系统的振动原理对船舶振动分析甚为重要。如果称主系统$M_1$的振动为总振动，从系统$M_2$的振动为局部振动，则以上的分析表明，当局部振动的质量$M_2$远小于总振动的质量$M_1$时，或局部振动的固有频率与总振动频率相差甚远时，可不计两者的耦合作用。因而，在计算船体总振动的最初几个固有频率时，常不考虑船体局部振动（如双层底等）的影响。船舶摇摆和船体振动可分别计算的原因也在于前者的固有频率比后者小一个数量级之故（摇摆频率的数量级为$0.1$次$/s$,而船体振动第一阶频率的数量级约为$1$次/$s$）。
