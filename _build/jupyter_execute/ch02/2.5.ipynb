# 动力吸振器的基本原理

某机器及其安装系统由 $M_1$ 和 $K_1$ 的单自由度系统作为其简化模型。如果在机器运转时，由于不平衡力 $F_0\sin\omega t$ 的作用，当激励频率 $\omega$ 接近系统固有频率时，主从系统便发生共振。如果该系统的参数以及机器运转速度都无法改变时，则可在此主系统上附加一个 $K_2$-$M_2$ 的从系统，适当选取从系统的质量和刚度，可以使主系统的振动大为减小，而附加的从系统却振动起来，就好像将主系统的振动能量都通过从系统吸收掉一样，这种附加的质量-弹簧系统便称为**动力吸振器**。其构成系统见{numref}`fig:2-19`，只需在 $M_1$ 上加上作用力 $F_0\sin\omega t$ 即可。

取重力作用下的静平衡位置为坐标原点，其运动方程式为：

$$
M_1\ddot{x}_1+(K_1+K_2)x_1-K_2x_2=F_0\sin\omega t
$$ (eq:2-133)

$$
M_2\ddot{x}_2+K_2x_2-K_2x_1=0
$$ (eq:2-134)

设其稳态响应为

$$
\begin{cases}
x_1=A_1\sin\omega t \\
x_2=A_2\sin\omega t
\end{cases}
$$ (eq:2-135)

由此得到

$$
A_1=\frac{(K_2-M_2\omega^2)F_0}{(K_1+K_2-M_1\omega^2)(K_2-M_2\omega^2)-K_2^2}
$$ (eq:2-136)

$$
A_2=\frac{K_2F_0}{(K_1+K_2-M_1\omega^2)(K_2-M_2\omega^2)-K_2^2}
$$ (eq:2-137)

由式{eq}`eq:2-136`可见，当 $\omega^2=\frac{K_2}{M_2}$ 时，主系统质量的振幅 $A_1$ 为零，也就是说，只要动力吸振器本身的固有频率等于激励频率时，就可以消除主系统质量的振动。此时，从系统质量的振幅

$$
A_2=-\frac{F_0}{K_2}
$$ (eq:2-138)

因此，从系统通过弹簧 $K_2$ 将一个大小与作用力 $F_0$ 相等，而方向却反向的作用力加到主质量 $M_1$ 上，正好与主系统上的激励力相抵消，从而消除主系统的振动，这便是动力吸振器的消振机理。

{numref}`fig:2-20`是对某个特定的主从系统动力响应的频响特性，其从系统与主系统的质量比 $a=0.05$，且 $\omega_{02}=\omega_{01}$。图中横坐标为激励频率与吸振器（即从系统）的固有频率之比 $\omega/\omega_{02}$，纵坐标表示主系统振幅与激励引起主质量的静伸长之比，即 $A_1/(F_0/K_1)$。

由图中可以看出，当 $\omega=\omega_{02}$ 时，$A_1=0$，即主系统不振动。图中实线表示有吸振器的两自由度系统的频率响应曲线，虚线表示无吸振器的主系统本身的幅频响应曲线。显然，附加吸振器后，系统出现两个新的共振频率 $\omega_{\mathrm{n}1}$ 与 $\omega_{\mathrm{n}2}$（由式(2.129)确定）。因此，当激励频率 $\omega$ 等于 $\omega_{\mathrm{n}1}$ 或 $\omega_{\mathrm{n}2}$ 时，系统将产生新的共振。为了消除主系统的振动，同时又不要产生新的共振，使主系统能够安全地工作在远离新的共振点的频率范围内，两个固有频率 $\omega_{\mathrm{n}1}$ 与 $\omega_{\mathrm{n}2}$ 以相距较远为好，为此需要适当选取 $a$ 值来实现。但若该吸振器质量与主系统质量之比过小时，为了吸收主系统的共振能量，吸振器质量的振幅 $A_2$ 将是一个较大值。这样，吸振器的弹簧刚度与运动位置必须满足该条件，在吸振器设计时，还需注意这一方面问题。

在此，我们再引入“反共振”的概念。主从系统中对于主系统的质量 $M_1$ 施加简谐激振力，在某个频率点——反共振频率点上，主系统虽然受到激励但是却保持不动，这就是“反共振”的现象。实际上，从力平衡的角度分析，因为该“反共振频率”等于从系统的固有频率，由于从系统 $M_2$ 共振运动的惯性力通过弹簧作用在 $M_1$ 上，抵消了外界的激振力从而达到平衡，好像主系统没有受激励一样处于静止状态。设计动力吸振的原理也就是利用反共振来实现。这个原理可以引申到连续系统，对于连续系统出现的反共振现象，同样是由于惯性力和外界激振力平衡的结果，解释的道理是一样的，但是，在做试验的时候，响应信号信噪比可能很差。

````python
from IPython.display import HTML, display

html = r"""
<script src="https://cdn.plot.ly/plotly-2.35.2.min.js"></script>
<style>
  #absorber-demo input[type="range"] {
    width: 100%;
    accent-color: #b91c1c;
  }
  #absorber-demo button {
    padding: 0.58rem 0.8rem;
    border-radius: 999px;
    border: 1px solid rgba(185,28,28,0.22);
    background: #ffffff;
    color: #0f172a;
    font-weight: 800;
    cursor: pointer;
  }
</style>
<div id="absorber-demo" style="border:1px solid rgba(185,28,28,0.14); border-radius:20px; padding:1rem; background:linear-gradient(180deg,#fef2f2 0%,#ffffff 100%);">
  <div style="display:grid; grid-template-columns:repeat(6,minmax(0,1fr)); gap:0.75rem; margin-bottom:0.95rem;">
    <div style="padding:0.78rem 0.86rem; border-radius:14px; background:#fff; border:1px solid rgba(148,163,184,0.18);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">质量比</div>
      <div id="ab-a" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#b91c1c;">a = 0.05</div>
    </div>
    <div style="padding:0.78rem 0.86rem; border-radius:14px; background:#fff; border:1px solid rgba(148,163,184,0.18);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">调谐比</div>
      <div id="ab-tune" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#b91c1c;">ω₀₂/ω₀₁ = 1.00</div>
    </div>
    <div style="padding:0.78rem 0.86rem; border-radius:14px; background:#fff; border:1px solid rgba(148,163,184,0.18);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">扫频比</div>
      <div id="ab-r" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#0f172a;">ω/ω₀₁ = 1.00</div>
    </div>
    <div style="padding:0.78rem 0.86rem; border-radius:14px; background:#fff; border:1px solid rgba(148,163,184,0.18);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">主系统幅值</div>
      <div id="ab-a1" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#0f172a;">A₁* = -</div>
    </div>
    <div style="padding:0.78rem 0.86rem; border-radius:14px; background:#fff; border:1px solid rgba(148,163,184,0.18);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">吸振器幅值</div>
      <div id="ab-a2" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#0f172a;">A₂* = -</div>
    </div>
    <div style="padding:0.78rem 0.86rem; border-radius:14px; background:#fff; border:1px solid rgba(148,163,184,0.18);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">当前状态</div>
      <div id="ab-state" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#0f172a;">-</div>
    </div>
  </div>

  <div style="display:grid; grid-template-columns:minmax(0,1fr); gap:1rem; align-items:start;">
    <div style="padding:0.95rem; border-radius:18px; background:#fff; border:1px solid rgba(148,163,184,0.18);">
      <div style="display:grid; grid-template-columns:repeat(3,minmax(0,1fr)); gap:0.78rem 0.9rem;">
        <div>
          <label style="display:flex; justify-content:space-between; gap:0.8rem; font-size:13px; color:#334155; font-weight:800;">
            <span>质量比 a = M₂ / M₁</span><span id="ab-a-label">0.05</span>
          </label>
          <input id="ab-a-slider" type="range" min="0.02" max="0.25" step="0.01" value="0.05">
        </div>
        <div>
          <label style="display:flex; justify-content:space-between; gap:0.8rem; font-size:13px; color:#334155; font-weight:800;">
            <span>调谐比 ω₀₂ / ω₀₁</span><span id="ab-tune-label">1.00</span>
          </label>
          <input id="ab-tune-slider" type="range" min="0.70" max="1.30" step="0.01" value="1.00">
        </div>
        <div>
          <label style="display:flex; justify-content:space-between; gap:0.8rem; font-size:13px; color:#334155; font-weight:800;">
            <span>扫频比 ω / ω₀₁</span><span id="ab-r-label">1.00</span>
          </label>
          <input id="ab-r-slider" type="range" min="0.40" max="1.80" step="0.01" value="1.00">
        </div>
      </div>
      <div style="display:flex; gap:0.5rem; flex-wrap:wrap; margin-top:0.95rem;">
        <button id="ab-tuned">精确调谐</button>
        <button id="ab-low">偏低调谐</button>
        <button id="ab-high">偏高调谐</button>
        <button id="ab-light">轻吸振器</button>
        <button id="ab-play">暂停动画</button>
      </div>
      <div style="margin-top:1rem; padding:0.85rem 0.9rem; border-radius:14px; background:#f8fafc; border:1px solid rgba(148,163,184,0.16);">
        <div style="font-size:13px; font-weight:800; color:#334155;">反共振解释</div>
        <div id="ab-note" style="margin-top:0.45rem; font-size:14px; line-height:1.5; color:#0f172a;">-</div>
      </div>
    </div>

    <div style="padding:0.9rem; border-radius:18px; background:#fff; border:1px solid rgba(148,163,184,0.18);">
      <canvas id="ab-canvas" width="820" height="260" style="width:100%; aspect-ratio:820 / 260; height:auto; display:block;"></canvas>
    </div>
    <div style="padding:0.95rem; border-radius:18px; background:#fff; border:1px solid rgba(148,163,184,0.18);">
      <div id="ab-response-plot" style="width:100%; height:360px;"></div>
    </div>
  </div>
</div>

<script>
(() => {
  const container = document.querySelector("#absorber-demo");
  if (!container || container.dataset.ready) return;
  container.dataset.ready = "true";

  const aSlider = container.querySelector("#ab-a-slider");
  const tuneSlider = container.querySelector("#ab-tune-slider");
  const rSlider = container.querySelector("#ab-r-slider");
  let playing = true;
  let phase = 0;
  let lastTime = null;
  let currentData = null;

  function amplitude(a, tune, r) {
    const M1 = 1;
    const K1 = 1;
    const F0 = 1;
    const M2 = a;
    const K2 = a * tune * tune;
    const den = (K1 + K2 - M1 * r * r) * (K2 - M2 * r * r) - K2 * K2;
    const A1 = Math.abs(den) < 1e-8 ? null : ((K2 - M2 * r * r) * F0) / den;
    const A2 = Math.abs(den) < 1e-8 ? null : (K2 * F0) / den;
    return {A1, A2, den, M2, K2};
  }

  function systemFrequencies(a, tune) {
    const xi = tune * tune;
    const disc = Math.max(0, (1 + (1 + a) * xi) ** 2 - 4 * xi);
    const z1 = (1 + (1 + a) * xi - Math.sqrt(disc)) / 2;
    const z2 = (1 + (1 + a) * xi + Math.sqrt(disc)) / 2;
    return {wn1: Math.sqrt(z1), wn2: Math.sqrt(z2), anti: tune};
  }

  function classify(r, freqs) {
    if (Math.abs(r - freqs.anti) < 0.03) return "反共振附近";
    if (Math.abs(r - freqs.wn1) < 0.05) return "第一共振附近";
    if (Math.abs(r - freqs.wn2) < 0.05) return "第二共振附近";
    if (r < freqs.wn1) return "低频区";
    return "过渡区";
  }

  function responseSeries(a, tune) {
    const freqs = systemFrequencies(a, tune);
    const rs = [];
    const withAbsorber = [];
    const withoutAbsorber = [];
    const absorberAmp = [];
    for (let r = 0.4; r <= 1.8001; r += 0.01) {
      const amp = amplitude(a, tune, r);
      rs.push(Number(r.toFixed(2)));
      const main = amp.A1 === null ? null : Math.min(25, Math.abs(amp.A1));
      const aux = amp.A2 === null ? null : Math.min(25, Math.abs(amp.A2));
      const bare = Math.abs(1 - r * r) < 1e-8 ? null : Math.min(25, Math.abs(1 / (1 - r * r)));
      withAbsorber.push(main);
      absorberAmp.push(aux);
      withoutAbsorber.push(bare);
    }
    return {freqs, rs, withAbsorber, withoutAbsorber, absorberAmp};
  }

  function update() {
    const a = Number(aSlider.value);
    const tune = Number(tuneSlider.value);
    const r = Number(rSlider.value);
    const amp = amplitude(a, tune, r);
    const series = responseSeries(a, tune);
    const state = classify(r, series.freqs);
    currentData = {a, tune, r, amp, series, state};

    container.querySelector("#ab-a-label").textContent = a.toFixed(2);
    container.querySelector("#ab-tune-label").textContent = tune.toFixed(2);
    container.querySelector("#ab-r-label").textContent = r.toFixed(2);
    container.querySelector("#ab-a").textContent = `a = ${a.toFixed(2)}`;
    container.querySelector("#ab-tune").textContent = `ω₀₂/ω₀₁ = ${tune.toFixed(2)}`;
    container.querySelector("#ab-r").textContent = `ω/ω₀₁ = ${r.toFixed(2)}`;
    container.querySelector("#ab-a1").textContent = amp.A1 === null ? "A₁* → ∞" : `A₁* = ${amp.A1.toFixed(3)}`;
    container.querySelector("#ab-a2").textContent = amp.A2 === null ? "A₂* → ∞" : `A₂* = ${amp.A2.toFixed(3)}`;
    container.querySelector("#ab-state").textContent = state;
    container.querySelector("#ab-note").textContent =
      state === "反共振附近"
        ? "此时吸振器的固有频率接近外激励频率，吸振器通过弹簧传回主质量的反向力与外激励几乎抵消，所以主质量几乎不动。"
        : state.includes("共振")
          ? "这里不是主系统振动被消掉，而是组合系统出现了新的共振峰。吸振器把原本单个共振峰分裂成了两个峰。"
          : "远离反共振点时，主系统与吸振器都会参与振动，但主系统的响应大小取决于调谐是否准确以及质量比是否足够。";

    Plotly.react(container.querySelector("#ab-response-plot"), [
      {x: series.rs, y: series.withAbsorber, mode: "lines", name: "有吸振器主系统", line: {color: "#b91c1c", width: 3}},
      {x: series.rs, y: series.withoutAbsorber, mode: "lines", name: "无吸振器主系统", line: {color: "#94a3b8", width: 2.2, dash: "dash"}},
      {x: series.rs, y: series.absorberAmp, mode: "lines", name: "吸振器幅值", line: {color: "#0f766e", width: 2.4}},
      {x: [r], y: [amp.A1 === null ? 25 : Math.min(25, Math.abs(amp.A1))], mode: "markers", name: "当前扫频点", marker: {size: 10, color: "#1d4ed8"}}
    ], {
      margin: {l: 52, r: 18, t: 18, b: 44},
      xaxis: {title: "ω / ω₀₁", range: [0.4, 1.8], autorange: false, fixedrange: true},
      yaxis: {title: "无因次幅值", range: [0, 25], autorange: false, fixedrange: true},
      legend: {x: 0.98, y: 0.98, xanchor: "right", yanchor: "top", bgcolor: "rgba(255,255,255,0.82)", bordercolor: "rgba(148,163,184,0.25)", borderwidth: 1},
      shapes: [
        {type: "line", x0: series.freqs.anti, x1: series.freqs.anti, y0: 0, y1: 25, line: {color: "rgba(15,118,110,0.35)", dash: "dot"}},
        {type: "line", x0: series.freqs.wn1, x1: series.freqs.wn1, y0: 0, y1: 25, line: {color: "rgba(185,28,28,0.25)", dash: "dash"}},
        {type: "line", x0: series.freqs.wn2, x1: series.freqs.wn2, y0: 0, y1: 25, line: {color: "rgba(185,28,28,0.25)", dash: "dash"}}
      ],
      paper_bgcolor: "rgba(0,0,0,0)",
      plot_bgcolor: "#ffffff"
    }, {displayModeBar: false, responsive: true});
  }

  function draw(scenePhase) {
    if (!currentData) return;
    const {a, tune, amp, state} = currentData;
    const canvas = container.querySelector("#ab-canvas");
    const ctx = canvas.getContext("2d");
    ctx.clearRect(0, 0, canvas.width, canvas.height);
    ctx.lineWidth = 3;
    ctx.strokeStyle = "#94a3b8";
    ctx.fillStyle = "#f8fafc";
    const labelFont = '700 24px "Microsoft YaHei", "PingFang SC", "Noto Sans SC", sans-serif';
    const infoFont = '600 18px "Microsoft YaHei", "PingFang SC", "Noto Sans SC", sans-serif';
    ctx.textBaseline = "middle";
    ctx.fillStyle = "#bfdbfe";
    ctx.fillRect(25, 44, 18, 156);
    ctx.strokeStyle = "#2563eb";
    ctx.strokeRect(25, 44, 18, 156);
    const y = 126;
    const center1 = 280;
    const center2 = 555;
    const A1 = amp.A1 === null ? 8.0 : amp.A1;
    const A2 = amp.A2 === null ? 8.0 : amp.A2;
    const maxAmp = Math.max(Math.abs(A1), Math.abs(A2), 0.8);
    const visualScale = maxAmp <= 1.2 ? 24 : 24 / Math.sqrt(maxAmp);
    const softLimit = 58;
    const softMap = (value) => softLimit * Math.tanh(value / softLimit);
    const x1 = center1 + softMap(visualScale * A1 * Math.sin(scenePhase));
    const x2 = center2 + softMap(visualScale * A2 * Math.sin(scenePhase));
    function spring(x0, x1p) {
      ctx.beginPath();
      ctx.moveTo(x0, y);
      for (let i = 1; i <= 18; i++) {
        const xp = x0 + (x1p - x0) * i / 19;
        const yp = y + (i % 2 ? -16 : 16);
        ctx.lineTo(xp, yp);
      }
      ctx.lineTo(x1p, y);
      ctx.stroke();
    }
    spring(43, x1 - 40);
    spring(x1 + 40, x2 - 40);
    [["主系统", x1, "#b91c1c"], ["吸振器", x2, "#0f766e"]].forEach(([label, x, color]) => {
      ctx.fillStyle = color;
      ctx.fillRect(x - 40, y - 32, 80, 64);
      ctx.fillStyle = "#fff";
      ctx.font = labelFont;
      ctx.textAlign = "center";
      ctx.fillText(label, x, y + 2);
    });
    ctx.fillStyle = "#0f172a";
    ctx.font = infoFont;
    ctx.textAlign = "left";
    ctx.fillText(`状态: ${state}`, 600, 52);
    ctx.fillText(`扫频: r = ${(Number(rSlider.value)).toFixed(2)}`, 600, 74);
    ctx.fillText(`M₂ = ${a.toFixed(2)}, K₂ = ${(a * tune * tune).toFixed(2)}`, 392, 220);
    ctx.fillText(`主系统位移: ${amp.A1 === null ? "很大" : amp.A1.toFixed(3)}`, 84, 220);
    ctx.fillText(`吸振器位移: ${amp.A2 === null ? "很大" : amp.A2.toFixed(3)}`, 84, 242);
    ctx.fillText(`动画缩放: ×${visualScale.toFixed(1)}（平滑压缩）`, 392, 242);
  }

  function applyPreset(a, tune, r) {
    aSlider.value = a.toFixed(2);
    tuneSlider.value = tune.toFixed(2);
    rSlider.value = r.toFixed(2);
    update();
  }

  container.querySelector("#ab-tuned").addEventListener("click", () => applyPreset(0.05, 1.00, 1.00));
  container.querySelector("#ab-low").addEventListener("click", () => applyPreset(0.05, 0.90, 1.00));
  container.querySelector("#ab-high").addEventListener("click", () => applyPreset(0.05, 1.10, 1.00));
  container.querySelector("#ab-light").addEventListener("click", () => applyPreset(0.03, 1.00, 1.00));
  container.querySelector("#ab-play").addEventListener("click", () => {
    playing = !playing;
    container.querySelector("#ab-play").textContent = playing ? "暂停动画" : "播放动画";
  });

  aSlider.addEventListener("input", update);
  tuneSlider.addEventListener("input", update);
  rSlider.addEventListener("input", update);

  function animate(timestamp) {
    if (lastTime === null) lastTime = timestamp;
    const dt = Math.min(0.05, (timestamp - lastTime) / 1000);
    lastTime = timestamp;
    if (playing) phase += dt * 3.0;
    draw(phase);
    requestAnimationFrame(animate);
  }

  update();
  requestAnimationFrame(animate);
})();
</script>
"""

display(HTML(html))
````
