# 隔振原理

## 第一类隔振

在船上有各种隔振，如主机、辅机等，它们在运转时所产生的干扰力会引起船体结构的振动。为了减少这些干扰力对船体的作用，通常在这些机械装置和船体结构之间装上一种称为减振器的隔振装置。它由弹性元件和阻尼元件组成。适当地选择减振器的弹簧常数和阻尼系数可使传递至船体结构的干扰力大为减小，这种隔振是把振源的干扰力和船体结构相隔离开，有时，称为第一类隔振，或简称为隔力。{numref}`fig:1-25`(a)是这种隔振的示意图。

```{figure} figures/fig1-25.png
---
name: fig:1-25
width: 60%
---
隔振示意图
```

对于第一类隔振，传递到基础的力是弹簧力和阻尼力之和，若采用复数解法，则传递力为

$$
F_{t}  = Kx + C\dot{x}
$$ (eq:1-122)

如果激励力是简谐的，$P\left ( t \right )  = P_{0}\sin \omega t$，则

$$
P\left ( t \right)= \bar{P} e^{i\omega t}, \bar{P} = P_{0} 
$$ (eq:1-123)

其稳态振动的解为

$$
x\left ( t \right) = \bar{X} e^{i\omega t}
$$ (eq:1-124)

式中，

$$
\bar{X} = \frac{\bar{P}}{K- M\omega ^{2}+ iC\omega} 
$$ (eq:1-125)

于是，通过弹簧和阻尼器传递到基础上的力为

$$
F_{t}  = Kx + C\dot{x}= \left ( K+  iC\omega\right )\bar{X}e^{i\omega t}=\bar{F_{t} }e^{i\omega t}
$$ (eq:1-126)

$$
\bar{F_{t} }= \left ( K+  iC\omega\right )\bar{X}= \frac{K+iC\omega}{K- M\omega ^{2}+iC\omega} \bar{P}  
$$ (eq:1-127)

式中，$\bar{F_{t} }$为传递力的复数形式的幅值。于是传递力与激励力的复数幅值比

$$
\frac{\bar{F_{t} }}{\bar{P}}  = \frac{K+iC\omega}{K- M\omega ^{2}+iC\omega}
$$ (eq:1-128)

由此，可求得传递力与激励力振幅值比（又称力传递函数）及相位差

$$
H_{F} \left ( \omega  \right )=  \frac{F_{t}}{P}= \frac{\sqrt{1+  \left ( 2\zeta r  \right )^{2}  } }{\sqrt{\left (  1-r ^{2} \right )^{2}+ \left ( 2\zeta r \right )^{2}   }   }
$$ (eq:1-129)

$$
\theta = \arctan \frac{2\zeta r ^{3} }{1- r ^{2}+  \left ( 2\zeta r  \right )^{2}   }
$$ (eq:1-130)

显然，它与1.6.1节中位移传递函数$H_{1} \left ( \omega  \right ) $是相同的。由{numref}`fig:1-18`可见，当$r = \sqrt{2} $时，$H_{F} \left ( \omega  \right )=1 $；当$r>  \sqrt{2}$时，$H_{F} \left ( \omega  \right )< 1$，即引起隔振的效果，$r$越大，$H_{F} \left ( \omega  \right )$越小，隔振效果越好。但$r$过大意味着$\omega _{n}$很小。这就造成隔振系统静伸长很大。在有外界低频骚扰时，往往会使动力机械不能正常工作。从{numref}`fig:1-18`还可看出当$r >  \sqrt{2}$时，阻尼增大反而使隔振的效果降低。因而隔振器的阻尼比不能设计得太大。

## 第二类隔振

为了把某些经不起振动的精密仪器等与船舶舱室内振动的环境隔离开来，使其受不到周围剧烈环境振动的影响，往往在这些仪器和固定它们的基础（如船体或船上的舱室）之间也要安装弹性隔振装置，有时这种隔振装置称第二类隔振，或简称隔幅，即隔去基础振幅之意。{numref}`fig:1-25`(b)是此种隔振的示意图。

对于第二类隔振，要求将被隔物体，即振动系统中质量$M$与外界基础或支座的振动隔离开，使被隔离的物体在空间保持静止状态。因此，人们关心的是其绝对位移的响应，其绝对位移传递函数的幅值及相位差同样可由式(1.88)和式(1.87)给出，与力的传递函数式{eq}`eq:1-129`和{eq}`eq:1-130`相同，因而隔振效果的结论也如第一类隔振所述。

隔振效果通常用隔振效率$\eta $表示，$\eta= \left [ 1- H_{F} \left ( \omega  \right ) \right ]\times100\%$。

由于实际的减振器阻尼比不大，故可近似地取$\zeta = 0 $时的$H_{F} \left ( \omega  \right )$ 

$$
H_{F} \left ( \omega  \right )= H_{1} \left ( \omega  \right )= \frac{1}{r ^{2}- 1 } =\frac{1}{\frac{4\pi ^{2} }{g}f^{2} \delta _{st}- 1  } 
$$ (eq:1-131)

式中，$f=\frac{\omega}{2\pi}$为隔振系统的输入频率；$\delta _{st}=\frac{Mg}{K}$为隔振系统的静伸长。

## 交互探索：第二类隔振如何减少设备绝对位移？

```{tip}
第二类隔振关注的是基础振动传到被隔物体上的位移。低频时设备几乎随基础一起运动；高频且 $r>\sqrt{2}$ 时，设备绝对位移显著小于基础位移。
```

````python
from IPython.display import display, HTML

html = r"""
<script src="https://cdn.plot.ly/plotly-2.35.2.min.js"></script>
<style>
  #motion-isolation-demo input[type="range"] {
    -webkit-appearance: none; appearance: none; width: 100%; height: 8px;
    border-radius: 999px; background: rgba(148, 163, 184, 0.34); outline: none;
  }
  #motion-isolation-demo input[type="range"]::-webkit-slider-thumb {
    -webkit-appearance: none; appearance: none; width: 20px; height: 20px;
    border-radius: 50%; background: #2563eb; border: 3px solid #ffffff;
    box-shadow: 0 4px 12px rgba(37, 99, 235, 0.28); cursor: pointer;
  }
</style>

<div id="motion-isolation-demo" style="
    max-width: 1080px; margin: 1.4rem auto; padding: 1.1rem;
    border-radius: 22px; background: linear-gradient(135deg, #f8fafc 0%, #eef6ff 100%);
    box-shadow: 0 18px 42px rgba(15, 23, 42, 0.09);
    border: 1px solid rgba(148, 163, 184, 0.20);
    font-family: 'Segoe UI', 'PingFang SC', 'Microsoft YaHei', sans-serif;
">
  <div style="display:grid; grid-template-columns:repeat(5, minmax(0, 1fr)); gap:0.75rem; margin-bottom:0.95rem;">
    <div style="padding:0.78rem 0.86rem; border-radius:14px; background:rgba(255,255,255,0.88); border:1px solid rgba(37,99,235,0.12);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">频率比</div>
      <div id="mi-r-value" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#1d4ed8;">r = 2.00</div>
    </div>
    <div style="padding:0.78rem 0.86rem; border-radius:14px; background:rgba(255,255,255,0.88); border:1px solid rgba(37,99,235,0.12);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">阻尼比</div>
      <div id="mi-zeta-value" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#1d4ed8;">ζ = 0.10</div>
    </div>
    <div style="padding:0.78rem 0.86rem; border-radius:14px; background:rgba(255,255,255,0.88); border:1px solid rgba(37,99,235,0.12);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">位移传递率</div>
      <div id="mi-tr-value" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#0f172a;">0.36</div>
    </div>
    <div style="padding:0.78rem 0.86rem; border-radius:14px; background:rgba(255,255,255,0.88); border:1px solid rgba(37,99,235,0.12);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">设备幅值 X/Y</div>
      <div id="mi-amp-value" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#0f172a;">0.36</div>
    </div>
    <div style="padding:0.78rem 0.86rem; border-radius:14px; background:rgba(255,255,255,0.88); border:1px solid rgba(37,99,235,0.12);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">状态</div>
      <div id="mi-state-value" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#0f172a;">隔振区</div>
    </div>
  </div>

  <div style="display:grid; grid-template-columns:repeat(2, minmax(0, 1fr)); gap:1.1rem; padding:0.35rem 0.55rem 1rem;">
    <div>
      <label for="mi-r-slider" style="display:flex; justify-content:space-between; gap:1rem; flex-wrap:wrap; font-size:14px; color:#334155; font-weight:800;">
        <span>调节频率比 r = ω / ω<sub>n</sub></span><span style="font-weight:500; color:#64748b;">0.10 ~ 4.00</span>
      </label>
      <input id="mi-r-slider" type="range" min="0.10" max="4.00" step="0.01" value="2.00">
    </div>
    <div>
      <label for="mi-zeta-slider" style="display:flex; justify-content:space-between; gap:1rem; flex-wrap:wrap; font-size:14px; color:#334155; font-weight:800;">
        <span>调节阻尼比 ζ</span><span style="font-weight:500; color:#64748b;">0.00 ~ 1.00</span>
      </label>
      <input id="mi-zeta-slider" type="range" min="0.00" max="1.00" step="0.05" value="0.10">
    </div>
  </div>

  <div style="display:grid; grid-template-columns:minmax(360px, 1fr) minmax(360px, 1fr); gap:1rem; align-items:stretch;">
    <div style="padding:0.9rem; border-radius:18px; background:#ffffff; border:1px solid rgba(148,163,184,0.18);">
      <svg id="mi-svg" viewBox="0 0 560 360" style="width:100%; height:360px; display:block; border-radius:14px; background:linear-gradient(180deg, #ffffff 0%, #f8fbff 100%);">
        <rect x="0" y="0" width="560" height="360" rx="14" fill="transparent"></rect>
        <line x1="80" y1="318" x2="500" y2="318" stroke="rgba(249,115,22,0.45)" stroke-width="2" stroke-dasharray="6 5"></line>
        <text x="30" y="306" fill="#f97316" style="font-size:15px; font-weight:800;">空间固定参考线</text>
        <rect id="mi-base" x="140" y="302" width="280" height="34" rx="8" fill="#93c5fd" stroke="#2563eb" stroke-width="3"></rect>
        <g id="mi-mass-group">
          <rect id="mi-mass" x="230" y="62" width="100" height="64" rx="12" fill="#60a5fa" stroke="#1d4ed8" stroke-width="3"></rect>
          <text x="248" y="101" fill="#1e3a8a" style="font-size:16px; font-weight:800;">仪器</text>
        </g>
        <polyline id="mi-spring" fill="none" stroke="#2563eb" stroke-width="4" stroke-linejoin="round" stroke-linecap="round"></polyline>
        <rect id="mi-damper-case" x="296" y="220" width="54" height="82" rx="8" fill="none" stroke="#94a3b8" stroke-width="4"></rect>
        <line id="mi-damper-piston" x1="300" y1="261" x2="346" y2="261" stroke="#64748b" stroke-width="5" stroke-linecap="round"></line>
        <line id="mi-damper-rod" x1="323" y1="126" x2="323" y2="261" stroke="#64748b" stroke-width="4" stroke-linecap="round"></line>
        <line id="mi-base-arrow" x1="462" y1="318" x2="462" y2="318" stroke="#2563eb" stroke-width="5" stroke-linecap="round"></line>
        <polygon id="mi-base-head" points="462,318 462,318 462,318" fill="#2563eb"></polygon>
        <line id="mi-mass-arrow" x1="462" y1="62" x2="462" y2="62" stroke="#f97316" stroke-width="5" stroke-linecap="round"></line>
        <polygon id="mi-mass-head" points="462,62 462,62 462,62" fill="#f97316"></polygon>
        <text x="480" y="346" fill="#2563eb" style="font-size:20px; font-weight:800;">y(t)</text>
        <text x="480" y="70" fill="#f97316" style="font-size:20px; font-weight:800;">x(t)</text>
      </svg>
    </div>
    <div style="padding:0.95rem; border-radius:18px; background:#ffffff; border:1px solid rgba(148,163,184,0.18);">
      <div id="mi-plot" style="width:100%; height:320px;"></div>
    </div>
  </div>

  <div style="margin-top:1rem; padding:0.95rem; border-radius:18px; background:#ffffff; border:1px solid rgba(148,163,184,0.18);">
    <div id="mi-time-plot" style="width:100%; height:310px;"></div>
  </div>
</div>

<script>
(function() {
  const container = document.getElementById("motion-isolation-demo");
  if (!container || container.dataset.initialized === "true") return;
  container.dataset.initialized = "true";

  const rSlider = container.querySelector("#mi-r-slider");
  const zetaSlider = container.querySelector("#mi-zeta-slider");
  const rValue = container.querySelector("#mi-r-value");
  const zetaValue = container.querySelector("#mi-zeta-value");
  const trValue = container.querySelector("#mi-tr-value");
  const ampValue = container.querySelector("#mi-amp-value");
  const stateValue = container.querySelector("#mi-state-value");
  const plotEl = container.querySelector("#mi-plot");
  const timePlot = container.querySelector("#mi-time-plot");
  const baseEl = container.querySelector("#mi-base");
  const massEl = container.querySelector("#mi-mass-group");
  const springEl = container.querySelector("#mi-spring");
  const damperPiston = container.querySelector("#mi-damper-piston");
  const damperRod = container.querySelector("#mi-damper-rod");
  const damperCase = container.querySelector("#mi-damper-case");
  const baseArrow = container.querySelector("#mi-base-arrow");
  const baseHead = container.querySelector("#mi-base-head");
  const massArrow = container.querySelector("#mi-mass-arrow");
  const massHead = container.querySelector("#mi-mass-head");
  let r = Number(rSlider.value);
  let zeta = Number(zetaSlider.value);
  let start = null;

  function transmissibility(freqRatio, dampingRatio) {
    return Math.sqrt(1 + (2 * dampingRatio * freqRatio) ** 2) /
      Math.sqrt((1 - freqRatio * freqRatio) ** 2 + (2 * dampingRatio * freqRatio) ** 2);
  }
  function phaseLag(freqRatio, dampingRatio) {
    let angle = Math.atan2(2 * dampingRatio * freqRatio ** 3, 1 - freqRatio ** 2 + (2 * dampingRatio * freqRatio) ** 2);
    if (angle < 0) angle += Math.PI;
    return angle;
  }
  function stateText(tr, freqRatio) {
    if (Math.abs(freqRatio - Math.sqrt(2)) < 0.05) return "临界边界";
    if (tr < 1) return "隔振区";
    if (freqRatio < Math.sqrt(2)) return "随动/放大区";
    return "接近边界";
  }
  function springPoints(baseY, massY) {
    const yTop = massY + 64;
    const yBottom = baseY;
    const x = 250;
    const points = [];
    const turns = 7;
    const amp = 24;
    points.push(`${x},${yTop}`);
    const usable = Math.max(yBottom - yTop, 30);
    const step = usable / (turns * 2);
    for (let i = 0; i < turns * 2; i += 1) {
      points.push(`${x + (i % 2 === 0 ? -amp : amp)},${yTop + step * (i + 1)}`);
    }
    points.push(`${x},${yBottom}`);
    return points.join(" ");
  }
  function displacementArrow(lineEl, headEl, x, y0, yTarget) {
    const delta = yTarget - y0;
    const len = Math.min(96, Math.abs(delta));
    const y2 = y0 + Math.sign(delta || 1) * len;
    lineEl.setAttribute("x1", x.toFixed(2));
    lineEl.setAttribute("x2", x.toFixed(2));
    lineEl.setAttribute("y1", y0.toFixed(2));
    lineEl.setAttribute("y2", y2.toFixed(2));
    if (len < 0.8) {
      headEl.setAttribute("points", `${x},${y0} ${x},${y0} ${x},${y0}`);
      return;
    }
    if (y2 < y0) {
      headEl.setAttribute("points", `${x},${y2} ${x - 10},${y2 + 16} ${x + 10},${y2 + 16}`);
    } else {
      headEl.setAttribute("points", `${x},${y2} ${x - 10},${y2 - 16} ${x + 10},${y2 - 16}`);
    }
  }
  function renderTransferPlot() {
    const xs = [];
    const curves = [0.02, 0.10, 0.30, 0.60].map((z) => ({ z, y: [] }));
    for (let i = 0; i <= 390; i += 1) {
      const x = 0.1 + i * 3.9 / 390;
      xs.push(x);
      curves.forEach((curve) => curve.y.push(Math.min(8, transmissibility(x, curve.z))));
    }
    const data = curves.map((curve) => ({
      x: xs,
      y: curve.y,
      mode: "lines",
      name: `ζ=${curve.z.toFixed(2)}`,
      line: { width: curve.z === 0.10 ? 3 : 2 },
      hovertemplate: "r=%{x:.2f}<br>H=%{y:.3f}<extra></extra>",
    }));
    data.push({
      x: [r],
      y: [Math.min(8, transmissibility(r, zeta))],
      mode: "markers",
      name: "当前点",
      marker: { size: 12, color: "#f97316", line: { color: "white", width: 2 } },
      hovertemplate: "r=%{x:.2f}<br>H=%{y:.3f}<extra></extra>",
    });
    data.push({
      x: [Math.sqrt(2), Math.sqrt(2)],
      y: [0, 8],
      mode: "lines",
      name: "r=√2",
      line: { color: "rgba(239,68,68,0.55)", width: 2, dash: "dash" },
      hoverinfo: "skip",
    });
    Plotly.react(plotEl, data, {
      margin: { l: 58, r: 18, t: 18, b: 52 },
      paper_bgcolor: "rgba(0,0,0,0)",
      plot_bgcolor: "#ffffff",
      xaxis: { title: "<b>频率比 r</b>", range: [0.1, 4], showgrid: true, gridcolor: "rgba(148,163,184,0.22)" },
      yaxis: { title: "<b>位移传递率 H<sub>X</sub></b>", range: [0, 4], showgrid: true, gridcolor: "rgba(148,163,184,0.22)" },
      legend: { orientation: "h", y: 1.02, x: 0.98, xanchor: "right", yanchor: "bottom" },
    }, { displayModeBar: false, responsive: true });
  }
  function renderTimePlot() {
    const t = [];
    const y = [];
    const x = [];
    const tr = transmissibility(r, zeta);
    const lag = phaseLag(r, zeta);
    const maxT = Math.max(10, 8 * 2 * Math.PI / Math.max(r, 0.1));
    for (let i = 0; i <= 800; i += 1) {
      const time = i * maxT / 800;
      t.push(time);
      y.push(Math.sin(r * time));
      x.push(tr * Math.sin(r * time - lag));
    }
    const yLimit = Math.max(1.2, Math.min(4, tr * 1.15));
    Plotly.react(timePlot, [
      { x: t, y, mode: "lines", name: "基础位移 y(t)", line: { color: "#2563eb", width: 2.6 }, hovertemplate: "τ=%{x:.2f}<br>y=%{y:.3f}<extra></extra>" },
      { x: t, y: x, mode: "lines", name: "设备绝对位移 x(t)", line: { color: "#f97316", width: 2.8 }, hovertemplate: "τ=%{x:.2f}<br>x=%{y:.3f}<extra></extra>" },
    ], {
      margin: { l: 58, r: 18, t: 18, b: 52 },
      paper_bgcolor: "rgba(0,0,0,0)",
      plot_bgcolor: "#ffffff",
      xaxis: { title: "<b>无量纲时间 τ = ω<sub>n</sub>t</b>", range: [0, maxT], showgrid: true, gridcolor: "rgba(148,163,184,0.22)" },
      yaxis: { title: "<b>位移幅值比</b>", range: [-yLimit, yLimit], showgrid: true, gridcolor: "rgba(148,163,184,0.22)" },
      legend: { orientation: "h", y: 1.02, x: 0.98, xanchor: "right", yanchor: "bottom" },
    }, { displayModeBar: false, responsive: true });
  }
  function updateReadouts() {
    const tr = transmissibility(r, zeta);
    rValue.textContent = `r = ${r.toFixed(2)}`;
    zetaValue.textContent = `ζ = ${zeta.toFixed(2)}`;
    trValue.textContent = tr.toFixed(3);
    ampValue.textContent = tr.toFixed(3);
    stateValue.textContent = stateText(tr, r);
    stateValue.style.color = tr < 1 ? "#1d4ed8" : "#b91c1c";
  }
  function animate(ts) {
    if (start === null) start = ts;
    const time = (ts - start) / 1000;
    const tr = transmissibility(r, zeta);
    const lag = phaseLag(r, zeta);
    const baseMotion = 32 * Math.sin(2 * Math.PI * 0.55 * time);
    const massMotion = Math.max(-110, Math.min(110, 32 * tr * Math.sin(2 * Math.PI * 0.55 * time - lag)));
    const baseY = 318 + baseMotion;
    const massY = 62 + massMotion;
    baseEl.setAttribute("y", baseY.toFixed(2));
    massEl.setAttribute("transform", `translate(0 ${massY - 62})`);
    const springBottomY = baseY;
    springEl.setAttribute("points", springPoints(springBottomY, massY));
    const caseY = baseY - 82;
    const relativeMotion = (massY - 62) - (baseY - 318);
    const pistonY = Math.max(caseY + 14, Math.min(caseY + 68, caseY + 41 + 0.45 * relativeMotion));
    damperCase.setAttribute("y", caseY.toFixed(2));
    damperPiston.setAttribute("y1", pistonY.toFixed(2));
    damperPiston.setAttribute("y2", pistonY.toFixed(2));
    damperRod.setAttribute("y1", (massY + 64).toFixed(2));
    damperRod.setAttribute("y2", pistonY.toFixed(2));
    displacementArrow(baseArrow, baseHead, 462, 318, baseY);
    displacementArrow(massArrow, massHead, 462, 62, massY);
    requestAnimationFrame(animate);
  }
  function apply() {
    r = Number(rSlider.value);
    zeta = Number(zetaSlider.value);
    updateReadouts();
    renderTransferPlot();
    renderTimePlot();
  }
  rSlider.addEventListener("input", apply);
  zetaSlider.addEventListener("input", apply);
  apply();
  requestAnimationFrame(animate);
})();
</script>
"""

display(HTML(html))
````
