在求解这个问题之前，先来看一种特殊情况。

一个初始静止的单自由度系统，在初始时刻受到一个单位脉冲$\delta \left ( t \right )$的作用，求其运动规律。

$\delta \left ( t \right )$可看做在一个非常短暂的时间间隔$\mu$内作用一个常量力$\frac{1}{\mu} $。于是该系统的运动方程为

$$
\left\{\begin{matrix}
m\ddot{x}\left ( t \right )+  c\dot{x}\left ( t \right )+  kx \left ( t \right )=\delta \left ( t \right ) 
\\  
x \left ( 0 \right )= 0;\dot{x}\left ( 0 \right )= 0
\end{matrix}\right. 
$$ (eq:1-99)

应用动量定理$m\bigtriangleup \upsilon = p\bigtriangleup t= \frac{1}{\mu} \mu= 1$，可以将初始时刻的单位脉冲转化为在$t= \mu $时刻的速度，即$\bigtriangleup \upsilon = \frac{1}{m}$。

因$\mu$非常微小，所以系统在$t= \mu $时可以视为仍处于初始状态，于是运动方程{eq}`eq:1-99`转化为以下形式

$$
\left\{\begin{matrix}
m\ddot{x}\left ( t \right )+  c\dot{x}\left ( t \right )+  kx \left ( t \right )=0 
\\  
x \left ( 0 \right )= 0;\dot{x}\left ( 0 \right )= \frac{1}{m} 
\end{matrix}\right. 
$$ (eq:1-100)

这是一个初始位移为$0$，初始速度为$\frac{1}{m} $的自由振动问题，它的解记作$x\left ( t \right )=h\left ( t \right ) $ ，由式{eq}`eq:1-38`可求得 

$$
\left\{\begin{matrix}
h\left ( t \right )=   \frac{1}{\omega _{\varepsilon } } e^{- \zeta\omega _{0}t\sin\omega _{\varepsilon }t  }   \left ( t\ge 0 \right )  
\\  
0\left ( t<  0 \right )  
\end{matrix}\right.
$$ (eq:1-101)

式中，$\omega _{0} = \sqrt{\frac{k}{m} } $、$\zeta  =\frac{c}{2\sqrt{mk} }  $分别为系统的自振圆频率与阻尼比；$\omega _{\varepsilon }  =\omega _{0} \sqrt{1-\zeta ^{2} } $为系统有阻尼自振圆频率。

$ h\left ( t \right ) $是在初始时刻作用单位脉冲而使单自由度系统产生的响应，所以称之为脉冲响应函数（Impulse Response Function）。

在时间间隔$\left [ 0,t \right ] $内作用的任意外载$F_{e} \left ( t  \right ) $，可以看做是在该时间间隔内的大量脉冲$F_{e} \left (\tau   \right ) \mathrm{d}\tau $的叠加，如{numref}`fig:1-24`所示。由每一个这种冲量而使系统产生的响应是$F_{e} \left (\tau   \right ) \mathrm{d}\tau h\left ( t- \tau  \right ) $它仅在时刻$\tau $之后才起作用。所以力$F_{e} \left ( t  \right ) $在时刻$t$的总响应为

```{figure} figures/fig1-24.png
---

---

```

$$
x\left ( t \right ) = \int_{0}^{t} F_{e} \left (\tau   \right )  h\left ( t- \tau  \right )\mathrm{d}\tau
$$ (eq:1-102)

由于在小于$0$的时间没有力作用，在大于$t$的时间力还没有作用上去，因此以上积分的上下限可以改为$+  \infty$和$-  \infty$，亦即

$$
x\left ( t \right ) = \int_{-  \infty }^{+  \infty } F_{e} \left (\tau   \right )  h\left ( t- \tau  \right )\mathrm{d}\tau
$$ (eq:1-103)
 
作变量代换$\theta = t- \tau $可得

$$
x\left ( t \right ) = \int_{-  \infty }^{+  \infty } F_{e} \left (t- \theta    \right )  h\left (\theta   \right )\mathrm{d}\theta  
$$ (eq:1-104)

式{eq}`eq:1-102`~式{eq}`eq:1-104`称为杜阿梅尔积分（Duhamel's Integral）或卷积积分（Convolution Integral），它们都是计算结构瞬态响应十分有用的工具。

事实上，如果在$t< 0$时就已经有力作用，式{eq}`eq:1-103`或式{eq}`eq:1-104`仍然是成立的，而式{eq}`eq:1-102`却不能用了。

一般初始条件下，单自由度系统的响应可表示为

$$
x\left ( t \right ) =e^{-\zeta \omega _{0}t  } \left ( \frac{\dot{x}_{0}+  \zeta \omega _{0} x_{0}   }{\omega _{\varepsilon } }\sin\omega _{\varepsilon }t+x_{0} \cos\omega _{\varepsilon }t      \right ) +   \int_{-  \infty }^{+  \infty } F_{e} \left (\tau   \right )  h\left ( t- \tau  \right )\mathrm{d}\tau
$$ (eq:1-105)

若不考虑阻尼，上式可退化为

$$
x\left ( t \right ) =\frac{\dot{x}_{0}}{\omega _{0} }    \sin\omega _{\varepsilon }t+x_{0} \cos\omega _{\varepsilon }t +   \int_{-  \infty }^{+  \infty } F_{e} \left (\tau   \right )  h\left ( t- \tau  \right )\mathrm{d}\tau
$$ (eq:1-106)

由以上讨论可知，任何外载$F_{e} \left ( t  \right ) $都可以通过划分为微小的短时间脉冲，借助于$h \left ( t  \right ) $在时间域的积分（即运用叠加原理）而获得系统的总响应。这种方法通常称为时域法（Time Domain Method）。时域法可用来计算任何线性单自由度系统对任意输入的响应。但有时用频域法（Frequency Domain Method）分析更为方便。频域分析在概念上和1.3节所介绍的周期荷载分析相似，即：首先把荷载展开为简谐作用分量，然后计算系统对每个作用分量的响应，最后再将各简谐分量响应叠加起来得到系统的总响应。数学上已经证明，对于一个时间函数$F_{e} \left ( t  \right ) $，如果积分$\int_{-  \infty }^{+  \infty }\left |  F_{e} \left (t \right ) \right |  \mathrm{d}t$为有限值［即$F_{e} \left ( t  \right ) $为绝对可积的］，则可将$F_{e} \left ( t  \right ) $变换为

$$
F_{e} \left ( t  \right ) = \frac{1}{2\pi} \int_{-\infty }^{+\infty}P\left ( \omega  \right )e^{i\omega t}  \mathrm{d}\omega
$$ (eq:1-107)

其中

$$
P\left ( \omega\right )=\int_{-\infty }^{+\infty}F_{e} \left ( t  \right )e^{- i\omega t}  \mathrm{d}t
$$ (eq:1-108)

式{eq}`eq:1-107`~式{eq}`eq:1-108`通常称为逆傅里叶变换（Inverse Fourier Transform）和正傅里叶变换（Direct Fourier Transform）。

与傅里叶级数表达式{eq}`eq:1-90`相类似，式{eq}`eq:1-107`可理解为将一个任意荷载用无穷个简谐分量的和来表示（傅里叶积分），式中的$\frac{P\left ( \omega\right )}{2\pi }$定义了频率为$\omega$处单位$\omega$的荷载分量的幅值。因此乘以频响函数$H\left ( \omega\right )$，就可得出频率为$\omega$时单位$\omega$的响应分量幅值。因此，在整个频率范围内对这些响应分量求和就可得到总的稳态响应，即

$$
x\left ( t \right )=\frac{1}{2\pi} \int_{-\infty }^{+\infty}H\left ( \omega\right )P\left ( \omega\right )e^{i\omega t}  \mathrm{d}\omega
$$ (eq:1-109)

为了应用频域法，必须借助式{eq}`eq:1-108`计算出给定荷载的简谐分量$P\left ( \omega\right )$以及由式{eq}`eq:1-94`或{eq}`eq:1-95`给出的频响函数。只有当荷载函数能进行傅里叶积分变换时，才能有效地利用频域分析方法。即便如此，计算最后所得的积分也可能是非常麻烦的。在实际应用中，往往是基于离散傅里叶变换（Discrete Fourier Transform，DFT）的概念，采用数值方法进行计算。快速傅里叶变换（Fsst Fourier Transform，FFT）就是一种有效的方法，它的基本概念可在一般的数值方法教科书中查到。

式{eq}`eq:1-103`和式{eq}`eq:1-109`都可表示单自由度系统的稳态响应，二者是等价的，即

$$
\frac{1}{2\pi} \int_{-\infty }^{+\infty}H\left ( \omega\right )P\left ( \omega\right )e^{i\omega t}  \mathrm{d}\omega=\int_{-  \infty }^{+  \infty } F_{e} \left (\tau   \right )  h\left ( t- \tau  \right )\mathrm{d}\tau
$$ (eq:1-110)

由式{eq}`eq:1-108`，上式可变换为

$$
\frac{1}{2\pi} \int_{-\infty }^{+\infty}H\left ( \omega\right )\left [ \int_{-\infty }^{+\infty}F_{e} \left ( \tau   \right )e^{- i\omega \tau }  \mathrm{d}\tau \right ] e^{i\omega t}  \mathrm{d}\omega=\int_{-  \infty }^{+  \infty } F_{e} \left (\tau   \right )  h\left ( t- \tau  \right )\mathrm{d}\tau
$$ (eq:1-111)

进一步化为

$$
\int_{-\infty }^{+\infty}F_{e} \left ( \tau   \right )\left [ \frac{1}{2\pi} \int_{-\infty }^{+\infty}H\left ( \omega\right )e^{i\omega  \left ( t -\tau  \right ) }   \mathrm{d}\omega \right ] \mathrm{d}\tau =\int_{-  \infty }^{+  \infty } F_{e} \left (\tau   \right )  h\left ( t- \tau  \right )\mathrm{d}\tau
$$ (eq:1-112)

令$u=t-\tau $，由上式可看出

$$
h\left ( u\right ) =\frac{1}{2\pi}\int_{-\infty }^{+\infty}  H\left ( \omega\right ) e^{i\omega u}\mathrm{d}\omega
$$ (eq:1-113)

即

$$
h\left ( t\right ) =\frac{1}{2\pi}\int_{-\infty }^{+\infty}  H\left ( \omega\right ) e^{i\omega t}\mathrm{d}\omega
$$ (eq:1-114)

在1.4节中，单自由度系统在单位复荷载$e^{i\omega t}$作用下的稳态响应为

$$
x\left ( t \right ) =H\left ( \omega\right ) e^{i\omega t}
$$ (eq:1-115)

这个单位复荷载激励起的结构稳态响应也可借助杜阿梅尔积分来计算，即
  
$$
x\left ( t \right ) = \int_{-  \infty }^{+  \infty } e^{i\omega \left (t-\theta   \right )}  h\left (\theta   \right )\mathrm{d}\theta 
$$ (eq:1-116)

比较式{eq}`eq:1-115`与式{eq}`eq:1-116`可知

$$
H\left ( \omega\right ) e^{i\omega t}= \int_{-  \infty }^{+  \infty } e^{i\omega \left (t-\theta   \right )}  h\left (\theta   \right )\mathrm{d}\theta 
$$ (eq:1-117)

将两边的公共因子$e^{i\omega t}$消去，就得到

$$
H\left ( \omega\right ) = \int_{-  \infty }^{+  \infty } h\left (\theta   \right ) e^{-i\omega \theta } \mathrm{d}\theta 
$$ (eq:1-118)

即

$$
H\left ( \omega\right ) = \int_{-  \infty }^{+  \infty } h\left (t \right ) e^{-i\omega t } \mathrm{d}t 
$$ (eq:1-119)

由式{eq}`eq:1-114`和式{eq}`eq:1-119`可知，脉冲响应函数$h\left (t \right )$和频率响应函数$H\left ( \omega\right )$之间构成傅氏变换对。

````python
from IPython.display import display, HTML

html = r"""
<script src="https://cdn.plot.ly/plotly-2.35.2.min.js"></script>
<style>
  #arbitrary-forced-demo input[type="range"] {
    -webkit-appearance: none;
    appearance: none;
    width: 100%;
    height: 8px;
    border-radius: 999px;
    background: rgba(148, 163, 184, 0.34);
    outline: none;
  }
  #arbitrary-forced-demo input[type="range"]::-webkit-slider-thumb {
    -webkit-appearance: none;
    appearance: none;
    width: 20px;
    height: 20px;
    border-radius: 50%;
    background: #dc2626;
    border: 3px solid #ffffff;
    box-shadow: 0 4px 12px rgba(220, 38, 38, 0.25);
    cursor: pointer;
  }
  #arbitrary-forced-demo select {
    width: 100%;
    box-sizing: border-box;
    border: 1px solid rgba(148, 163, 184, 0.35);
    border-radius: 12px;
    padding: 0.55rem 0.65rem;
    color: #0f172a;
    background: #ffffff;
    font-weight: 800;
  }
</style>

<div id="arbitrary-forced-demo" style="
    max-width: 1120px; margin: 1.4rem auto; padding: 1.1rem;
    border-radius: 22px; background: linear-gradient(135deg, #fff7f7 0%, #f6fbff 100%);
    box-shadow: 0 18px 42px rgba(15, 23, 42, 0.09);
    border: 1px solid rgba(148, 163, 184, 0.20);
    font-family: 'Segoe UI', 'PingFang SC', 'Microsoft YaHei', sans-serif;
">
  <div style="display:grid; grid-template-columns:repeat(5, minmax(0, 1fr)); gap:0.75rem; margin-bottom:0.95rem;">
    <div style="padding:0.78rem 0.85rem; border-radius:14px; background:rgba(255,255,255,0.88); border:1px solid rgba(220,38,38,0.12);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">阻尼比</div>
      <div id="af-zeta-value" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#b91c1c;">ζ = 0.10</div>
    </div>
    <div style="padding:0.78rem 0.85rem; border-radius:14px; background:rgba(255,255,255,0.88); border:1px solid rgba(220,38,38,0.12);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">荷载类型</div>
      <div id="af-type-value" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#0f172a;">矩形脉冲</div>
    </div>
    <div style="padding:0.78rem 0.85rem; border-radius:14px; background:rgba(255,255,255,0.88); border:1px solid rgba(220,38,38,0.12);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">最大激励</div>
      <div id="af-fmax-value" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#0f172a;">1.00</div>
    </div>
    <div style="padding:0.78rem 0.85rem; border-radius:14px; background:rgba(255,255,255,0.88); border:1px solid rgba(220,38,38,0.12);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">最大响应</div>
      <div id="af-xmax-value" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#0f172a;">0.00</div>
    </div>
    <div style="padding:0.78rem 0.85rem; border-radius:14px; background:rgba(255,255,255,0.88); border:1px solid rgba(220,38,38,0.12);">
      <div style="font-size:12px; color:#64748b; font-weight:700;">贡献脉冲时刻</div>
      <div id="af-probe-value" style="margin-top:0.25rem; font-size:18px; font-weight:800; color:#0f172a;">τ = 2.0</div>
    </div>
  </div>

  <div style="display:grid; grid-template-columns:repeat(3, minmax(0, 1fr)); gap:1rem; align-items:end;">
    <div>
      <label for="af-load-type" style="font-size:14px; color:#334155; font-weight:800;">荷载形式</label>
      <select id="af-load-type" style="margin-top:0.35rem;">
        <option value="rect">矩形脉冲</option>
        <option value="tri">三角脉冲</option>
        <option value="half">半正弦脉冲</option>
        <option value="multi">多脉冲</option>
        <option value="random">随机样荷载</option>
      </select>
    </div>
    <div>
      <label for="af-zeta-slider" style="display:flex; justify-content:space-between; gap:1rem; flex-wrap:wrap; font-size:14px; color:#334155; font-weight:800;">
        <span>阻尼比 ζ</span><span style="font-weight:500; color:#64748b;">0.02 ~ 0.50</span>
      </label>
      <input id="af-zeta-slider" type="range" min="0.02" max="0.50" step="0.01" value="0.10">
    </div>
    <div>
      <label for="af-probe-slider" style="display:flex; justify-content:space-between; gap:1rem; flex-wrap:wrap; font-size:14px; color:#334155; font-weight:800;">
        <span>查看某一脉冲贡献 τ</span><span style="font-weight:500; color:#64748b;">0 ~ 20</span>
      </label>
      <input id="af-probe-slider" type="range" min="0" max="20" step="0.05" value="2.0">
    </div>
  </div>

  <div style="display:grid; grid-template-columns:repeat(3, minmax(0, 1fr)); gap:1rem; margin-top:0.9rem;">
    <div>
      <label for="af-amp-slider" style="display:flex; justify-content:space-between; gap:1rem; flex-wrap:wrap; font-size:14px; color:#334155; font-weight:800;">
        <span>幅值 A</span><span id="af-amp-label" style="font-weight:500; color:#64748b;">1.00</span>
      </label>
      <input id="af-amp-slider" type="range" min="0.2" max="3.0" step="0.05" value="1.0">
    </div>
    <div>
      <label for="af-start-slider" style="display:flex; justify-content:space-between; gap:1rem; flex-wrap:wrap; font-size:14px; color:#334155; font-weight:800;">
        <span>开始时间 τ<sub>0</sub></span><span id="af-start-label" style="font-weight:500; color:#64748b;">2.0</span>
      </label>
      <input id="af-start-slider" type="range" min="0" max="12" step="0.1" value="2.0">
    </div>
    <div>
      <label for="af-duration-slider" style="display:flex; justify-content:space-between; gap:1rem; flex-wrap:wrap; font-size:14px; color:#334155; font-weight:800;">
        <span>持续时间 Δτ</span><span id="af-duration-label" style="font-weight:500; color:#64748b;">3.0</span>
      </label>
      <input id="af-duration-slider" type="range" min="0.3" max="8.0" step="0.1" value="3.0">
    </div>
  </div>

  <div style="display:grid; grid-template-columns:minmax(0, 1fr); gap:1rem; margin-top:1rem;">
    <div style="padding:0.95rem; border-radius:18px; background:#ffffff; border:1px solid rgba(148,163,184,0.18);">
      <div id="af-impulse-plot" style="width:100%; height:175px;"></div>
    </div>
  </div>

  <div style="display:grid; grid-template-columns:minmax(360px, 1.2fr) minmax(300px, 0.8fr); gap:1rem; margin-top:1rem; align-items:stretch;">
    <div style="padding:0.95rem; border-radius:18px; background:#ffffff; border:1px solid rgba(148,163,184,0.18);">
      <div id="af-force-plot" style="width:100%; height:250px;"></div>
    </div>
    <div style="padding:0.9rem; border-radius:18px; background:#ffffff; border:1px solid rgba(148,163,184,0.18);">
      <svg id="af-oscillator" viewBox="0 0 560 220" style="width:100%; height:175px; display:block; border-radius:14px; background:linear-gradient(180deg, #ffffff 0%, #fff7f7 100%);">
        <rect x="0" y="0" width="560" height="220" rx="14" fill="transparent"></rect>
        <rect x="22" y="38" width="18" height="140" rx="5" fill="#94a3b8"></rect>
        <line x1="40" y1="96" x2="60" y2="96" stroke="#94a3b8" stroke-width="3" stroke-linecap="round"></line>
        <line x1="60" y1="84" x2="300" y2="84" stroke="#94a3b8" stroke-width="3" stroke-linecap="round"></line>
        <line x1="60" y1="108" x2="300" y2="108" stroke="#94a3b8" stroke-width="3" stroke-linecap="round"></line>
        <line x1="60" y1="84" x2="60" y2="108" stroke="#94a3b8" stroke-width="3" stroke-linecap="round"></line>
        <line id="af-damper-piston" x1="104" y1="86" x2="104" y2="106" stroke="#64748b" stroke-width="4" stroke-linecap="round"></line>
        <line id="af-damper-rod" x1="104" y1="96" x2="178" y2="96" stroke="#64748b" stroke-width="4" stroke-linecap="round"></line>
        <line x1="40" y1="142" x2="520" y2="142" stroke="rgba(148,163,184,0.32)" stroke-width="6" stroke-linecap="round"></line>
        <line x1="340" y1="32" x2="340" y2="188" stroke="rgba(249,115,22,0.42)" stroke-width="2" stroke-dasharray="6 5"></line>
        <polyline id="af-spring" fill="none" stroke="#dc2626" stroke-width="4" stroke-linejoin="round" stroke-linecap="round"></polyline>
        <rect id="af-mass" x="306" y="83" width="68" height="68" rx="12" fill="#fca5a5" stroke="#b91c1c" stroke-width="3"></rect>
        <line id="af-force-arrow" x1="392" y1="70" x2="450" y2="70" stroke="#ef4444" stroke-width="5" stroke-linecap="round"></line>
        <polygon id="af-force-head" points="450,70 438,62 438,78" fill="#ef4444"></polygon>
        <text x="292" y="204" fill="#64748b" style="font-size:13px;">平衡位置</text>
        <text id="af-force-label" x="392" y="54" fill="#ef4444" style="font-size:20px; font-weight:700;">F(t)</text>
      </svg>
      <div style="display:flex; gap:0.65rem; margin-top:0.75rem;">
        <button id="af-play" style="flex:1; padding:0.78rem 0.75rem; border:none; border-radius:999px; background:#dc2626; color:white; font-weight:800; cursor:pointer;">播放</button>
        <button id="af-reset" style="flex:1; padding:0.78rem 0.75rem; border:1px solid rgba(220,38,38,0.22); border-radius:999px; background:white; color:#0f172a; font-weight:800; cursor:pointer;">重播</button>
      </div>
    </div>
  </div>

  <div style="display:grid; grid-template-columns:minmax(0, 1fr); gap:1rem; margin-top:1rem;">
    <div style="padding:0.95rem; border-radius:18px; background:#ffffff; border:1px solid rgba(148,163,184,0.18);">
      <div id="af-response-plot" style="width:100%; height:360px;"></div>
    </div>
    <div style="padding:0.95rem; border-radius:18px; background:#ffffff; border:1px solid rgba(148,163,184,0.18);">
      <div id="af-spectrum-plot" style="width:100%; height:320px;"></div>
    </div>
  </div>
</div>

<script>
(function() {
  const container = document.getElementById("arbitrary-forced-demo");
  if (!container || container.dataset.initialized === "true") return;
  container.dataset.initialized = "true";

  const tauMax = 20;
  const steps = 1200;
  const dt = tauMax / steps;
  const playbackDuration = 13000;
  const omegaN = 1;
  const mass = 1;
  const equilibriumX = 340;
  const massWidth = 68;
  const baseMotionScale = 55;
  const maxVisualTravel = 120;
  const damperRodLength = 132;

  const loadTypeEl = container.querySelector("#af-load-type");
  const zetaSlider = container.querySelector("#af-zeta-slider");
  const ampSlider = container.querySelector("#af-amp-slider");
  const startSlider = container.querySelector("#af-start-slider");
  const durationSlider = container.querySelector("#af-duration-slider");
  const probeSlider = container.querySelector("#af-probe-slider");
  const zetaValue = container.querySelector("#af-zeta-value");
  const typeValue = container.querySelector("#af-type-value");
  const fmaxValue = container.querySelector("#af-fmax-value");
  const xmaxValue = container.querySelector("#af-xmax-value");
  const probeValue = container.querySelector("#af-probe-value");
  const ampLabel = container.querySelector("#af-amp-label");
  const startLabel = container.querySelector("#af-start-label");
  const durationLabel = container.querySelector("#af-duration-label");
  const playBtn = container.querySelector("#af-play");
  const resetBtn = container.querySelector("#af-reset");
  const forcePlot = container.querySelector("#af-force-plot");
  const impulsePlot = container.querySelector("#af-impulse-plot");
  const responsePlot = container.querySelector("#af-response-plot");
  const spectrumPlot = container.querySelector("#af-spectrum-plot");
  const springEl = container.querySelector("#af-spring");
  const massEl = container.querySelector("#af-mass");
  const pistonEl = container.querySelector("#af-damper-piston");
  const rodEl = container.querySelector("#af-damper-rod");
  const arrowEl = container.querySelector("#af-force-arrow");
  const headEl = container.querySelector("#af-force-head");
  const forceLabel = container.querySelector("#af-force-label");

  let zeta = Number(zetaSlider.value);
  let loadType = loadTypeEl.value;
  let amplitude = Number(ampSlider.value);
  let startTime = Number(startSlider.value);
  let duration = Number(durationSlider.value);
  let probeTime = Number(probeSlider.value);
  let response = computeAll();
  let currentIndex = 0;
  let isPlaying = false;
  let startTimestamp = null;
  let rafId = null;

  function loadTypeLabel(type) {
    return {
      rect: "矩形脉冲",
      tri: "三角脉冲",
      half: "半正弦脉冲",
      multi: "多脉冲",
      random: "随机样荷载",
    }[type] || "矩形脉冲";
  }

  function pseudoRandom(seed) {
    const value = Math.sin(seed * 12.9898) * 43758.5453;
    return value - Math.floor(value);
  }

  function forceAt(t) {
    const s = startTime;
    const d = Math.max(duration, dt);
    if (loadType === "rect") {
      return (t >= s && t <= s + d) ? amplitude : 0;
    }
    if (loadType === "tri") {
      if (t < s || t > s + d) return 0;
      const local = (t - s) / d;
      return amplitude * (local <= 0.5 ? local * 2 : (1 - local) * 2);
    }
    if (loadType === "half") {
      if (t < s || t > s + d) return 0;
      return amplitude * Math.sin(Math.PI * (t - s) / d);
    }
    if (loadType === "multi") {
      const centers = [s, s + 0.55 * d, s + 1.18 * d];
      const widths = [0.18 * d, 0.22 * d, 0.16 * d];
      const scales = [1.0, -0.72, 0.52];
      return centers.reduce((sum, center, idx) => {
        const dist = Math.abs(t - center);
        if (dist > widths[idx]) return sum;
        return sum + amplitude * scales[idx] * (1 - dist / widths[idx]);
      }, 0);
    }
    if (loadType === "random") {
      if (t < s || t > s + d) return 0;
      const local = (t - s) / d;
      const envelope = Math.sin(Math.PI * local);
      const k = Math.floor(local * 36);
      const a = pseudoRandom(k + 3) * 2 - 1;
      const b = pseudoRandom(k + 4) * 2 - 1;
      const frac = local * 36 - k;
      const noise = a * (1 - frac) + b * frac;
      return amplitude * envelope * noise;
    }
    return 0;
  }

  function impulseResponse(t) {
    if (t < 0) return 0;
    if (zeta < 1) {
      const omegaD = omegaN * Math.sqrt(1 - zeta * zeta);
      return Math.exp(-zeta * omegaN * t) * Math.sin(omegaD * t) / (mass * omegaD);
    }
    const eps = 1e-4;
    const omegaD = omegaN * Math.sqrt(1 - (1 - eps) * (1 - eps));
    return Math.exp(-(1 - eps) * omegaN * t) * Math.sin(omegaD * t) / (mass * omegaD);
  }

  function discreteSpectrum(values) {
    const sampleCount = 256;
    const stride = Math.max(1, Math.floor(values.length / sampleCount));
    const sampled = [];
    for (let i = 0; i < values.length && sampled.length < sampleCount; i += stride) sampled.push(values[i]);
    const n = sampled.length;
    const freqs = [];
    const mags = [];
    const maxK = 48;
    for (let k = 0; k <= maxK; k += 1) {
      let re = 0;
      let im = 0;
      for (let j = 0; j < n; j += 1) {
        const angle = -2 * Math.PI * k * j / n;
        re += sampled[j] * Math.cos(angle);
        im += sampled[j] * Math.sin(angle);
      }
      freqs.push(2 * Math.PI * k / tauMax);
      mags.push(Math.sqrt(re * re + im * im) / n);
    }
    return { freqs, mags };
  }

  function computeAll() {
    const tau = [];
    const force = [];
    const h = [];
    for (let i = 0; i <= steps; i += 1) {
      const t = i * dt;
      tau.push(t);
      force.push(forceAt(t));
      h.push(impulseResponse(t));
    }

    const x = new Array(force.length).fill(0);
    for (let i = 0; i < force.length; i += 1) {
      let sum = 0;
      for (let j = 0; j <= i; j += 1) {
        sum += force[j] * h[i - j] * dt;
      }
      x[i] = sum;
    }

    const probeIndex = Math.max(0, Math.min(force.length - 1, Math.round(probeTime / dt)));
    const contribution = tau.map((t, i) => (
      i >= probeIndex ? force[probeIndex] * dt * h[i - probeIndex] : null
    ));
    const maxForce = Math.max(0.001, ...force.map((value) => Math.abs(value)));
    const maxX = Math.max(0.001, ...x.map((value) => Math.abs(value)));
    const motionScale = Math.min(baseMotionScale, maxVisualTravel / maxX);
    const forceSpectrum = discreteSpectrum(force);
    const responseSpectrum = discreteSpectrum(x);
    return { tau, force, h, x, contribution, probeIndex, maxForce, maxX, motionScale, forceSpectrum, responseSpectrum };
  }

  function buildSpringPoints(startX, endX, y) {
    const points = [];
    const straight = 18;
    const turns = 10;
    const amplitude = 16;
    points.push(`${startX},${y}`);
    points.push(`${startX + straight},${y}`);
    const usable = Math.max(endX - startX - straight * 2, 28);
    const step = usable / (turns * 2);
    for (let i = 0; i < turns * 2; i += 1) {
      const x = startX + straight + step * (i + 1);
      points.push(`${x},${y + (i % 2 === 0 ? -amplitude : amplitude)}`);
    }
    points.push(`${endX - straight},${y}`);
    points.push(`${endX},${y}`);
    return points.join(" ");
  }

  function baseLayout(yTitle, yRange) {
    return {
      margin: { l: 68, r: 18, t: 18, b: 52 },
      paper_bgcolor: "rgba(0,0,0,0)",
      plot_bgcolor: "#ffffff",
      xaxis: {
        title: "<b>无量纲时间 τ = ω<sub>n</sub>t</b>",
        range: [0, tauMax],
        showgrid: true,
        gridcolor: "rgba(148, 163, 184, 0.22)",
        zeroline: false,
        showline: true,
        linecolor: "rgba(15, 23, 42, 0.25)",
      },
      yaxis: {
        title: yTitle,
        range: yRange,
        showgrid: true,
        gridcolor: "rgba(148, 163, 184, 0.22)",
        zeroline: true,
        zerolinecolor: "rgba(15, 23, 42, 0.20)",
      },
      hoverlabel: { bgcolor: "#ffffff", bordercolor: "rgba(220,38,38,0.30)", font: { color: "#0f172a" } },
      showlegend: false,
    };
  }

  function renderPlots(index, fullRedraw = false) {
    const partialTau = response.tau.slice(0, index + 1);
    const partialForce = response.force.slice(0, index + 1);
    const partialX = response.x.slice(0, index + 1);
    const fLimit = Math.max(1, response.maxForce * 1.18);
    const hLimit = Math.max(0.6, Math.max(...response.h.map((value) => Math.abs(value))) * 1.18);
    const xLimit = Math.max(0.6, response.maxX * 1.18);

    if (fullRedraw) {
      Plotly.react(forcePlot, [
        {
          x: response.tau,
          y: response.force,
          mode: "lines",
          line: { color: "rgba(15,23,42,0.20)", width: 2 },
          hovertemplate: "τ = %{x:.2f}<br>F = %{y:.3f}<extra></extra>",
        },
        {
          x: partialTau,
          y: partialForce,
          mode: "lines",
          line: { color: "#dc2626", width: 3 },
          hovertemplate: "τ = %{x:.2f}<br>F = %{y:.3f}<extra></extra>",
        },
        {
          x: [response.tau[response.probeIndex], response.tau[response.probeIndex]],
          y: [-fLimit, fLimit],
          mode: "lines",
          line: { color: "#f97316", width: 2, dash: "dot" },
          hoverinfo: "skip",
        },
      ], baseLayout("<b>任意激励 F(τ)</b>", [-fLimit, fLimit]), { displayModeBar: false, responsive: true });

      Plotly.react(impulsePlot, [
        {
          x: response.tau,
          y: response.h,
          mode: "lines",
          line: { color: "#334155", width: 2.8 },
          hovertemplate: "τ = %{x:.2f}<br>h = %{y:.4f}<extra></extra>",
        },
      ], baseLayout("<b>脉冲响应函数 h(τ)</b>", [-hLimit, hLimit]), { displayModeBar: false, responsive: true });

      Plotly.react(responsePlot, [
        {
          x: response.tau,
          y: response.x,
          mode: "lines",
          name: "总响应",
          line: { color: "rgba(15,23,42,0.20)", width: 2 },
          hovertemplate: "τ = %{x:.2f}<br>x = %{y:.4f}<extra></extra>",
        },
        {
          x: response.tau,
          y: response.contribution,
          mode: "lines",
          name: "单脉冲贡献",
          line: { color: "#f97316", width: 2, dash: "dot" },
          hovertemplate: "τ = %{x:.2f}<br>贡献 = %{y:.4f}<extra></extra>",
        },
        {
          x: partialTau,
          y: partialX,
          mode: "lines",
          name: "已绘响应",
          line: { color: "#dc2626", width: 3 },
          hovertemplate: "τ = %{x:.2f}<br>x = %{y:.4f}<extra></extra>",
        },
        {
          x: [response.tau[index]],
          y: [response.x[index]],
          mode: "markers",
          marker: { size: 10, color: "#16f929", line: { color: "white", width: 2 } },
          name: "当前点",
          hovertemplate: "τ = %{x:.2f}<br>x = %{y:.4f}<extra></extra>",
        },
      ], {
        ...baseLayout("<b>响应 x(τ)</b>", [-xLimit, xLimit]),
        showlegend: true,
        legend: { orientation: "h", y: 1.02, x: 0.98, xanchor: "right", yanchor: "bottom" },
      }, { displayModeBar: false, responsive: true });

      renderSpectrumPlot();
    } else {
      Plotly.restyle(forcePlot, { x: [partialTau], y: [partialForce] }, [1]);
      Plotly.restyle(responsePlot, { x: [partialTau], y: [partialX] }, [2]);
      Plotly.restyle(responsePlot, { x: [[response.tau[index]]], y: [[response.x[index]]] }, [3]);
    }
  }

  function renderSpectrumPlot() {
    Plotly.react(spectrumPlot, [
      {
        x: response.forceSpectrum.freqs,
        y: response.forceSpectrum.mags,
        mode: "lines",
        name: "|P(ω)|",
        line: { color: "rgba(220,38,38,0.78)", width: 2.6 },
        hovertemplate: "ω/ωn = %{x:.2f}<br>|P| = %{y:.4f}<extra></extra>",
      },
      {
        x: response.responseSpectrum.freqs,
        y: response.responseSpectrum.mags,
        mode: "lines",
        name: "|X(ω)|",
        line: { color: "rgba(37,99,235,0.78)", width: 2.6 },
        hovertemplate: "ω/ωn = %{x:.2f}<br>|X| = %{y:.4f}<extra></extra>",
      },
    ], {
      margin: { l: 68, r: 18, t: 18, b: 52 },
      paper_bgcolor: "rgba(0,0,0,0)",
      plot_bgcolor: "#ffffff",
      xaxis: {
        title: "<b>频率比 ω / ω<sub>n</sub></b>",
        range: [0, 8],
        showgrid: true,
        gridcolor: "rgba(148, 163, 184, 0.22)",
        zeroline: false,
        showline: true,
        linecolor: "rgba(15, 23, 42, 0.25)",
      },
      yaxis: {
        title: "<b>幅值谱</b>",
        showgrid: true,
        gridcolor: "rgba(148, 163, 184, 0.22)",
        zeroline: true,
        zerolinecolor: "rgba(15, 23, 42, 0.20)",
      },
      legend: { orientation: "h", y: 1.02, x: 0.98, xanchor: "right", yanchor: "bottom" },
      hoverlabel: { bgcolor: "#ffffff", bordercolor: "rgba(220,38,38,0.30)", font: { color: "#0f172a" } },
    }, { displayModeBar: false, responsive: true });
  }

  function updateOscillator(index) {
    const x = response.x[index];
    const force = response.force[index];
    const massCenterX = equilibriumX + x * response.motionScale;
    const massLeft = massCenterX - massWidth / 2;
    const pistonX = massLeft - damperRodLength;
    massEl.setAttribute("x", massLeft.toFixed(2));
    springEl.setAttribute("points", buildSpringPoints(40, massLeft + 2, 142));
    pistonEl.setAttribute("x1", pistonX.toFixed(2));
    pistonEl.setAttribute("x2", pistonX.toFixed(2));
    rodEl.setAttribute("x1", pistonX.toFixed(2));
    rodEl.setAttribute("x2", massLeft.toFixed(2));

    const arrowStart = massLeft + massWidth + 18;
    const arrowLength = Math.min(62, 48 * Math.abs(force) / Math.max(1, response.maxForce));
    const arrowEnd = arrowStart + (force >= 0 ? arrowLength : -arrowLength);
    arrowEl.setAttribute("x1", arrowStart.toFixed(2));
    arrowEl.setAttribute("x2", arrowEnd.toFixed(2));
    if (force >= 0) {
      headEl.setAttribute("points", `${arrowEnd},70 ${arrowEnd - 12},62 ${arrowEnd - 12},78`);
    } else {
      headEl.setAttribute("points", `${arrowEnd},70 ${arrowEnd + 12},62 ${arrowEnd + 12},78`);
    }
    forceLabel.setAttribute("x", Math.min(500, Math.max(48, arrowStart)).toFixed(2));
  }

  function updateReadouts() {
    zetaValue.textContent = `ζ = ${zeta.toFixed(2)}`;
    typeValue.textContent = loadTypeLabel(loadType);
    fmaxValue.textContent = response.maxForce.toFixed(2);
    xmaxValue.textContent = response.maxX.toFixed(3);
    probeValue.textContent = `τ = ${probeTime.toFixed(2)}`;
    ampLabel.textContent = amplitude.toFixed(2);
    startLabel.textContent = startTime.toFixed(1);
    durationLabel.textContent = duration.toFixed(1);
  }

  function render(index, fullRedraw = false) {
    currentIndex = index;
    updateReadouts();
    updateOscillator(index);
    renderPlots(index, fullRedraw);
  }

  function recompute(fullRedraw = true) {
    response = computeAll();
    currentIndex = Math.min(currentIndex, response.tau.length - 1);
    render(0, fullRedraw);
  }

  function stop() {
    isPlaying = false;
    startTimestamp = null;
    if (rafId) cancelAnimationFrame(rafId);
    rafId = null;
    playBtn.textContent = "播放";
  }

  function tick(timestamp) {
    if (!isPlaying) return;
    if (startTimestamp === null) {
      startTimestamp = timestamp - (currentIndex / steps) * playbackDuration;
    }
    const progress = Math.min((timestamp - startTimestamp) / playbackDuration, 1);
    const nextIndex = Math.min(Math.floor(progress * steps), steps);
    if (nextIndex !== currentIndex) render(nextIndex, false);
    if (progress >= 1) {
      stop();
      return;
    }
    rafId = requestAnimationFrame(tick);
  }

  function play(fromStart = false) {
    if (fromStart) render(0, true);
    isPlaying = true;
    startTimestamp = null;
    playBtn.textContent = "暂停";
    rafId = requestAnimationFrame(tick);
  }

  function applyControlChanges() {
    zeta = Number(zetaSlider.value);
    loadType = loadTypeEl.value;
    amplitude = Number(ampSlider.value);
    startTime = Number(startSlider.value);
    duration = Number(durationSlider.value);
    probeTime = Number(probeSlider.value);
    stop();
    recompute(true);
    play(false);
  }

  [loadTypeEl, zetaSlider, ampSlider, startSlider, durationSlider, probeSlider].forEach((el) => {
    el.addEventListener("input", applyControlChanges);
    el.addEventListener("change", applyControlChanges);
  });

  playBtn.addEventListener("click", () => {
    if (isPlaying) stop();
    else play(currentIndex >= steps);
  });

  resetBtn.addEventListener("click", () => {
    stop();
    play(true);
  });

  render(0, true);
  play(false);
})();
</script>
"""

display(HTML(html))
````

```{note}
**例1-9** 一不计阻尼的单自由度系统，初始静止，受简谐扰力$F_{e} \left ( t  \right ) =P\cos \theta t$作用。试用杜阿梅尔积分求它的响应。
解：对于无阻尼系统，脉冲响应函数由式{eq}`eq:1-101`可写为

$$
\begin{aligned}
h\left ( t \right )= \left\{\begin{matrix}
\dfrac{1}{m\omega}\sin \omega t \left ( t\ge 0 \right )  \cr
0\left ( t<  0 \right )  
\end{matrix}\right.
\end{aligned} \tag{1}
$$

进而由式{eq}`eq:1-106`可求得系统的响应为

$$
\begin{aligned}
x\left ( t \right ) &= \int_{-  \infty }^{+  \infty } F_{e} \left (\tau   \right )  h\left ( t- \tau  \right )\mathrm{d}\tau= \frac{P}{m\omega } \int_{0}^{t} \cos \theta \tau \sin \omega \left ( t- \tau  \right )\mathrm{d}\tau \cr
&= \frac{P}{m\left ( \omega ^{2} - \theta ^{2}  \right ) }\cos \theta t- \frac{P}{m\left ( \omega ^{2} - \theta ^{2}  \right ) }\cos \omega  t
\end{aligned} \tag{2}
$$
```

应用傅里叶变换进行频域分析时，要求荷载函数$F_{e} \left ( t  \right ) $在$\left ( - \infty , +  \infty \right ) $内满足绝对可积等较强的条件。这样的条件，即使对于很简单的函数，如常数、多项式和三角函数等，都不能满足。拉普拉斯（Laplace）变换可以避免这样的困难。和傅里叶变换一样，拉普拉斯变换也是一种常用的研究线性系统的积分变换方法。它的严格数学定义和性质可参考有关的数学书籍，这里只简单介绍应用它来求解线性系统的动力响应。

设函数$f\left ( t  \right ) $满足条件：

1. 在$t\ge 0$的任一有限区间上分段连续可导；
2. 当$t\to+  \infty $时,$f\left ( t  \right ) $的增长速度不超过某个指数函数，亦即存在某个常数$M$及$\sigma _{0} \ge 0 $,使得$\left | f\left ( t  \right ) \right | \le Me^{\sigma _{0} t} \left ( 0< t< +  \infty  \right )$
其中，$\sigma _{0}  $称为$f\left ( t  \right ) $的增长指数。则称

$$
L\left ( p \right )  = \mathcal{L}\left [ f\left ( t  \right ) \right ] = \int_{0}^{+  \infty } f\left ( t  \right ) e^{- pt}\mathrm{d}t
$$ (eq:1-120)

为函数$f\left ( t  \right ) $的拉氏（拉普拉斯）变换，而称

$$
f\left ( t  \right ) = \mathcal{L} ^{- 1} \left [ L\left ( p \right )  \right ]= \frac{1}{2\pi i}\int_{\sigma - i\infty }^{\sigma +   i\infty}L\left ( p \right ) e^{pt}\mathrm{d}p
$$ (eq:1-121)

为复变函数$ L\left ( p \right )$的逆拉氏变换。拉普拉斯变换存在的条件要比傅里叶变换弱很多，描述工程问题的大多数荷载函数都能够满足。在实际工作中，简单函数的拉普拉斯变换可查找拉普拉斯变换表。

```{note}
**例1-10** 应用拉普拉斯变换求解单自由度系统在任意荷载$F_{e} \left ( t  \right )$作用下的响应。设初始条件为$x\left ( 0 \right ) = x_{0}$及$\dot{x} \left ( 0 \right ) = \dot{x}_{0} $。

解：单自由度系统在任意荷载$F_{e} \left ( t  \right )$作用下的运动方程为

$$
\ddot{x} \left ( t \right ) +  2\zeta \omega _{0} \dot{x}\left ( t \right )+  \omega _{0}^{2}x\left ( t \right )= \frac{1}{m} F_{e} \left ( t  \right ) \tag{1}
$$

将式(1)两边做拉普拉斯变换

$$
\mathcal{L} \left [ \ddot{x} \left ( t \right ) +  2\zeta \omega _{0} \dot{x}\left ( t \right )+  \omega _{0}^{2}x\left ( t \right ) \right ]= \frac{1}{m}\mathcal{L} \left [ F_{e} \left ( t  \right ) \right ]=\frac{1}{m}F\left ( p \right ) \tag{2}
$$

其中

$$
\mathcal{L} \left [ \dot{x}\left ( t \right )  \right ]=\int_{0}^{+\infty }e^{-pt}\dot{x}\left ( t \right )\mathrm{d}t=e^{-pt}x\left ( t \right ) \Bigg|_{0}^{+\infty}+  p\int_{0}^{+\infty}e^{-pt}x\left ( t \right ) \mathrm{d}t = pX\left ( p \right )- x_{0} \tag{3}
$$

同理

$$
\mathcal{L} \left [ \ddot{x}\left ( t \right )  \right]= p^{2}X\left ( p \right ) - px_{0} -\dot{x}_{0} \tag{4}
$$

因此，式(2)可写为

$$
X\left ( p \right )= \frac{F\left ( p \right )}{m\left ( p^{2}+  2\zeta \omega _{0}p+ \omega _{0}^{2}\right)}+\frac{p+2\zeta \omega _{0}}{p^{2}+  2\zeta \omega _{0}p+ \omega _{0}^{2}}x_{0}+\frac{1}{p^{2}+  2\zeta \omega _{0}p+ \omega _{0}^{2}}\dot{x}_{0} \tag{5}
$$

将上式右端化为简单的分式，然后利用卷积定理和拉氏变换表即可求出$X \left (p \right )$的拉普拉斯逆变换

$$
\begin{aligned}
x(t) = &\mathcal{L}^{-1}\left[ X(p) \right] = e^{-\zeta \omega_{0}t} \left( \frac{\dot{x}_{0} + \zeta \omega_{0} x_{0}}{\omega_{\varepsilon}} \sin\omega_{\varepsilon}t + x_{0} \cos\omega_{\varepsilon}t \right) + \cr
&\frac{1}{m\omega_{\varepsilon}} \int_{-\infty}^{+\infty} F_{e}(\tau) e^{-\zeta \omega_{0}(t - \tau)} \sin\omega_{\varepsilon}(t - \tau) \,\mathrm{d}\tau
\end{aligned} \tag{6}
$$

可以看出，式(6)与式{eq}`eq:1-105`是一致的。
```
